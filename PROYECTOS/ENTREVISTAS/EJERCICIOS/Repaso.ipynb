```python
### Ordenascon lambda y sorted
productos = [
    {"nombre": "A", "precio": 50,  "tags": ["x", "z"]},
    {"nombre": "B", "precio": 20,  "tags": ["z"]},
    {"nombre": "C", "precio": 100, "tags": ["x", "y", "z"]},
    {"nombre": "D", "precio": 20,  "tags": []}
]

### Sorted no sabe hacerlo descendnte
res = sorted(productos, key=lambda data: (data["precio"], -len(data["tags"])))
print("Ordenado por edad (ascendente):")
for p in res:
    print(p)
```

```python
class Cuenta:
    saldo = 0

    def __init__(self, nombre):
        self.nombre = nombre

    def depositar(self, cantidad):
        self.saldo += cantidad

c1 = Cuenta("Luis")
c2 = Cuenta("María")

c1.depositar(100)
print(c2.saldo)
```

Si implementas __bool__, Python usará ese método para saber si tu objeto es verdadero o falso.

Si no implementas __bool__, Python intenta usar __len__.

Si tampoco hay __len__, el objeto es siempre True.

```python
class Punto:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    ### Sobrecargar el operador suma
    def __add__(self, other):
        return Punto(self.x + other.x, self.y + other.y)

    def __repr__(self):
        return f"Punto({self.x}, {self.y})"

    def __bool__(self):
        return self.x != 0 and self.y != 0

p1 = Punto(2, 3)
p2 = Punto(5, 7)

print(p1 + p2)

p = Punto(3, 4)
if p:
    print("El punto es válido")
```

```python
#### MUTABILIDAD COMPARTIDA 
class Persona:
    compras = []

    def __init__(self, nombre):
        self.nombre = nombre

    def agregar(self, producto):
        self.compras.append(producto)

p1 = Persona("Luis")
p2 = Persona("María")

p1.agregar("Manzana")
p2.agregar("Pan")

print(p1.compras)
print(p2.compras)
```
