# Fundamentos del Lenguaje Python

### Semántica

*  Una característica importante del lenguaje Python es la consistencia de su modelo de objetos. Cada número, cadena de texto, estructura de datos, función, clase, módulo, etc. existe en el intérprete de Python en su propia «caja», lo que se denomina objeto Python.
*  Las funciones pueden admitir argumentos posicionales y de palabra clave: ` result = f(a, b, c, d=5, e=”foo”)`
*  En Python, a y b se refieren ahora en realidad al mismo objeto, la lista original [1, 2, 3]. Puede probarlo por sí mismo añadiendo un elemento a a y después examinando b

```python
a = [1, 2, 3]
b = a
print(b)
a.append(4)
print(b)
```

*  Cuando se pasan objetos como argumentos a una función, se crean nuevas variables locales que hacen referencia a los objetos originales sin copiar nada.
*  Por eso es posible modificar el interior de un argumento mutable. Supongamos que tenemos la siguiente función

```python
def append_element(some_list, element):
    some_list.append(element)

data = [1, 2, 3]
append_element(data, 4)
data
```

* A los atributos y métodos también se puede acceder mediante la función getattr:

```python
a = "foo"
getattr(a, "split")
```

* se le denomina «duck typing», que se podría traducir como «tipado de pato», aunque no tiene mucho sentido, porque se emplea esta expresión por el dicho «si camina como un pato y grazna como un pato, entonces es un pato».
*  es posible verificar que un objeto es iterable si implementa el protocolo iterador. Para muchos objetos, esto significa que tiene un «método mágico» `__iter__`, aunque una alternativa y mejor forma de comprobarlo es intentar usar la función iter

```python
def isiterable(obj):
 try:
     iter(obj)
     return True
 except TypeError: # no iterable
     return False

print(isiterable("a string"))
print(isiterable([1, 2, 3]))
print(isiterable(5))
```

### Operadores

 * a + b: Suma a y b
 * a – b: Resta b de a
 * a * b: Multiplica a por b
 * a / b: Divide a por b
 * a // b: División de piso de a por b, es decir, calcula el cociente de la división entre a y b, sin tener en cuenta el resto
 * a ** b: Eleva a a la potencia de b
 * a & b: True si tanto a como and b son True; para enteros, toma AND bitwise (o a nivel de bit)
 * a | b: True si a o b son True; para enteros, toma OR bitwise
 * a ^ b: Para booleanos, True si a o b es True, pero no ambos; para enteros, toma OR EXCLUSIVO bitwise
 * a == b : True si a es igual a b
 * a != b :  True si a no es igual a b
 * a < b, a <= b: True si a es menor (menor o igual) que b. Para booleanos, True si a o b es True, pero no ambos; para enteros, toma OR EXCLUSIVO bitwise
 * a > b, a >= b: True si a es mayor (mayor o igual) que b
 * a is b: True si a y b hacen referencia al mismo objeto Python
 * a is not b: True si a y b hacen referencia a distintos objetos Python

### Objetos mutables e inmutables

* Muchos objetos en Python, como listas, diccionarios, arrays NumPy y la mayoría de los tipos (clases) definidos por el usuario, son mutables. Esto significa que el objeto o los valores que contiene se pueden modificar.
*  Conviene recordar que, simplemente porque el hecho de que se pueda mutar un objeto, no significa que siempre se deba hacer.

```python
# Objeto mutable
a_list = ["foo", 2, [4, 5]]
a_list[2] = (3, 4)
a_list

# Objeto inmutable
a_tuple = (3, 5, (4, 5))
#a_tuple[1] = "four"
```

### Tipos Escalares

* Python tiene un pequeño conjunto de tipos integrados para manejar datos numéricos, cadenas de texto, valores booleanos (True o False) y fechas y horas. A estos tipos de “valores sencillos” se les denomina tipos escalares.
* Tipos de Escalares:
    * `None`: El valor «null» de Python (solo existe una instancia del objeto None)
    * `str`: Tipo cadena de texto; contiene textos Unicode
    * `bytes`: Datos binarios sin procesar
    * `float`: Un valor booleano True o False
    * `bool`: Número de punto flotante de precisión doble (observe que no existe un tipo double distinto)
    * `int`: Entero de precisión arbitraria

* Los principales tipos de Python para los números son int y float. Un int puede almacenar números arbitrariamente grandes.
* Los números de punto flotante se representan con el tipo float de Python. Internamente, cada uno es un valor de precisión doble. También se pueden expresar con notación científica.
* La división de enteros que no dé como resultado otro número entero siempre producirá un número de punto flotante.
* Para lograr una división de enteros al estilo de C (que no tiene en cuenta el resto si el resultado no es un número entero), utilizamos el operador de división de piso

```python
ival = 17239871
print(ival ** 6)
fval = 7.243
print(fval)
fval2 = 6.78e-5
print(fval2)

# Divisiones
print(3 / 2)
print(3 // 2)
```

### Cadenas de Texto

* Las cadenas de texto de Python son inmutables; no se pueden modificar. ` a = “this is a string ” `, `a[10] = “f”`
* Si necesitamos modificar una cadena de texto, tenemos que utilizar una función o un método que cree una nueva cadena, como el método `replace` para cadenas de texto.
* Muchos objetos de Python se pueden transformar en una cadena de texto utilizando la función `str`.
* Las cadenas de texto son una secuencia de caracteres Unicode y, por lo tanto, se les puede tratar igual que otras secuencias, como por ejemplo las listas y las tuplas.
* La sintaxis `s[:3]` se denomina `slicing` y se implementa para muchos tipos de secuencias de Python
* El carácter de la barra invertida `\` es un carácter de escape, lo que significa que se emplea para especificar caracteres especiales, como el carácter de línea nueva `\n`, o caracteres Unicode.
* Si tenemos una cadena de texto con muchas barras invertidas y ningún carácter especial, podría ser bastante molesto. Por suerte se puede poner delante de la primera comilla del texto una r, que significa que los caracteres se deben interpretar tal y como están. ***La r significa raw (sin procesar).***

```python
a = 'one way of writing a string'
b = "another way"
c = """
This is a longer string that
spans multiples lines
 """
print(a)
print(b)
print(c)
print(c.count("\n"))

# sustituir
b = a.replace("string", "longer string")
print(b)

# a string
a = 5.6
s = str(a)
print(s)

# separar en listas
s = "python"
list(s)
print(s[:3])

# Caracteres especiales
s = "12\\34"
print(s)

# raw
s = r"this\has\no\special\characters"
print(s)
```

*  Los objetos de cadena de texto tienen un método format que se puede utilizar para sustituir argumentos formateados dentro de la cadena, produciendo una nueva.
*  `template = "{0:.2f} {1:s} are worth US${2:d}"`: {0:.2f} significa formatear el primer argumento como un número de punto flotante con dos
 decimales. {1:s} significa formatear el segundo argumento como una cadena de texto. {2:d} significa formatear el tercer argumento como un entero exacto.
* Python 3.6 introdujo una nueva característica llamada ***«cadenas-f»*** (abreviatura de «literales de cadena formeateados») que puede lograr que sea aún más cómoda la creación de cadenas formateadas.
* `rate:.2f`: con este podemos escificar la cantidad de digitos.

```python
template = "{0:.2f} {1:s} are worth US${2:d}"
print(template.format(88.46, "Argentine Pesos", 1))
# cadena formateada con f
amount = 10
rate = 88.46
currency = "Pesos"
result = f"{amount} {currency} is worth US${amount / rate}"
print(result)

# Espeficiammos delimitadores de formato
result = f"{amount} {currency} is worth US${amount / rate:.2f}"
print(result)
```

### Bytes y UniCode

* Unicode se ha convertido en el tipo de cadena de texto de primer orden en permitir un manejo más sólido de texto ASCII y no ASCII.
* Las cadenas de texto eran todo bytes sin una codificación Unicode explícita.
* Podemos convertir esta cadena Unicode en su representación de bytes UTF-8 utilizando el método `encode`
*  Es más habitual encontrar objetos bytes cuando se trabaja con archivos, donde quizá no sea deseable decodificar implícitamente todos los datos a cadenas Unicode.

```python
# una cadena de texto Unicode de ejemplo con caracteres no ASCII
val = "español"
print("Cadena", val)

# usnado unicode
val_utf8 = val.encode("utf-8")
print("Hexadecimal:", val_utf8)

# imprimiendo tipos
print("Normal",type(val))
print("UTF-8",type(val_utf8))

# podemos volver atrás utilizando el método decode
print("Decode:",val_utf8.decode("utf-8"))
```

###  Conversión de tipos (Type casting)

*  Los dos valores booleanos de Python se escriben como True y False. Las comparaciones y otras expresiones condicionales evalúan a True o False. Los valores booleanos se combinan con las palabras clave ***and*** y ***or***.
* Cuando se convierten a números, False se convierte en 0 y True en 1.
* Los tipos `str`, `bool`, `int` y `float` son también funciones que se pueden usar para convertir valores a dichos tipos
* Observe que la mayoría de los valores que no son cero, cuando se convierten a bool, son True.
* None es el tipo de valor de Python nulo o null.
* None es también un valor predeteminado habitual para argumentos de función
*  El módulo `datetime` integrado de Python ofrece los tipos datetime, date y time. El tipo datetime combina la información almacenada en date y time y es el más utilizado
*  Dada una instancia datetime, se pueden extraer los objetos equivalentes date y time llamando a métodos de la datetime del mismo nombre
*  El método `strftime` formatea un datetime como cadena de texto
*   Las cadenas de texto se pueden convertir (analizar) en objetos datetime con la función `strptime`
*   Como datetime.datetime es un tipo inmutable, métodos como este siempre producen nuevos objetos

```python
print(int(False))
print(int(True))

# Cadena a flotante
s = "3.14159"
fval = float(s)
print("fval", type(fval))

# Flotante a entero (lo redondea)
print(int(fval))

# Entero a booleano
print(bool(fval))

# uso de None
a = None
print(a is None)
b = 5
print(b is not None)
def add_and_maybe_multiply(a, b, c=None):
    result = a + b
    if c is not None:
        result = result * c
    return result
print(add_and_maybe_multiply(1,2,3))
print(add_and_maybe_multiply(1,2))

# Fechas y horas
from datetime import datetime, date, time
dt = datetime(2011, 10, 29, 20, 30, 21)
print(dt.day)
print(dt.minute)
print(dt.date(),":", dt.time())
print(dt.date(),":", dt.time())
# Los importante son los % y el escape
print(dt.strftime("%Y-%m-%d %H:%M:%S"))
print(dt.strftime("%Y/%m/%d [%H:%M:%S]"))

# Podemos hacer caso contrario
print(datetime.strptime("20091031", "%Y%m%d"))
print(datetime.strptime("10/31/2009", "%m/%d/%Y"))

# Cambiando datos, pero como es inmutable, crea otro objeto
dt_hour = dt.replace(minute=7, second=5)
print(dt_hour)

# Diferencia de tiempos
dt2 = datetime(2011, 11, 15, 22, 30)
delta = dt2-dt
print(delta)
```

###  Control de Flujo

* Los bucles for se utilizan para aplicar determinadas instrucciones a una colección (como una lista o una tupla) o a un iterador.
* Se puede avanzar un bucle for a la siguiente repetición, saltando el resto del bloque, mediante la palabra clave continue.
* Un bucle for puede darse por terminado también con la palabra clave break.  La palabra clave break solo finaliza el bucle for más interno; los bucles for exteriores seguirán ejecutándose.
* si los elementos de la colección o iterador son secuencias (tuplas o listas, por ejemplo), se pueden desempaquetar cómodamente en variables de la sentencia del bucle for --> ***for a, b, c in iterator***
* Un bucle while especifica una condición y un bloque de código que se va ejecutar hasta que la condición evalúe a False o el bucle sea finalizado explícitamente con break.
* `pass` es la sentencia «no operativa» (es decir, que no hace nada) de Python. Se puede utilizar en los bloques en los que no hay que realizar ninguna acción (o como marcador para código que aún no se ha implementado); 

```python
# usamos el continue
sequence = [1, 2, None, 4, None, 5]
total = 0
for value in sequence:
    if value is None:
        continue
    total += value
print(total)

# Break hasta el 5.
sequence = [1, 2, 0, 4, 6, 5, 2, 1]
total_until_5 = 0
for value in sequence:
    if value == 5:
        break
    total_until_5 += value
print(total_until_5)

# Bucles anidados
for i in range(4):
    for j in range(4):
        if j > i:
            break
        print((i, j))

# Ciclo While
x = 256
total = 0
while x > 0:
    if total > 500:
        break
    total += x
    x = x // 2
```

### Range

*  La función range genera una secuencia de enteros espaciados por igual.
*  No incluyen el digito que muestra, ya que se cuenta desde cero.
*  Se suele utilizar para aplicar instrucciones a distintas secuencias según un índice
*  

```python
# Son 10 numeros separados de uno en uno desde el cero
lista= list(range(10))
print(lista)

# separados por 2 en 2
lista= list(range(0, 20, 2))
print(lista)

# De reversa
lista= list(range(5, 0, -1))
print(lista)

# recorrer una lista
seq = [1, 2, 3, 4]
for i in range(len(seq)):
    print(f"element {i}: {seq[i]}")
```
