# Funciones

* Las funciones son el método principal y más importante de Python para organizar y reutilizar código.
* Las funciones se declaran con la palabra clave `def`. Una función contiene un bloque de código, con un uso opcional de la palabra clave `return`.
* No hay problema por tener varias sentencias return. Si Python alcanza el final de una función sin encontrar una sentencia return, devuelve automáticamente `None`. 

### Espacios de nombres, ámbito y funciones locales

* Las funciones pueden acceder a variables creadas dentro de la misma función, así como a las que están fuera de la función en ámbitos más elevados (o incluso globales).
* Cualquier variable asignada dentro de una función de forma predeterminada está asignada asimismo al espacio de nombres local.
* Cuando se llama a la función, se crea el espacio de nombres, que se llena inmediatamente con los argumentos de la función. Una vez finalizada esta, el espacio de nombres local se destruye.
* Es posible asignar variables fuera del ámbito de la función, pero dichas variables deben estar declaradas explícitamente utilizando las palabras clave `global` o `nonlocal`.
* `nonlocal` permite a una función modificar variables definidas en un ámbito de mayor nivel que no es global.

```python
a = []
def func():
    for i in range(5):
        a.append(i)
func()
print(a)

a = None
def bind_a_variable():
    global a
    a = []

bind_a_variable()    
print(a)

# Non Local
def contador():
    numero = 0 
    def incrementar():
        nonlocal numero # Instaciamos la variable de arriba.
        numero += 1
        print("Dentro de incrementar:", numero)
    incrementar()
    return numero

resultado = contador()
print("Resultado final:", resultado)
```

### Devolver varios valores

* En análisis de datos y otras aplicaciones científicas se hace esto a menudo. Lo que está ocurriendo aquí es que la función devuelve realmente solo un objeto, una tupla, que después se desempaqueta para obtener las variables del resultado.
* En este caso, return_value sería una tupla de tres, conteniendo las tres variables devueltas.
* Una alternativa a devolver varios valores del modo que hemos visto, que quizá resulte interesante, sería devolver un diccionario.

```python
def f():
    a = 5
    b = 6
    c = 7
    return a, b, c

a, b, c = f()
print(a, b, c)
print(f()) # realmente regresa una tupla.

def b():
    a = 5
    b = 6
    c = 7
    return {"a" : a, "b" : b, "c" : c}
dicc = b()
print(dicc)
```

### Las funciones son objetos

* Las funciones Python son objetos, muchas construcciones que son difíciles de codificar en otros lenguajes se pueden expresar aquí fácilmente.
* Una forma de hacerlo es utilizando métodos de cadena de texto internos junto con el módulo de librería estándar re para las expresiones regulares.
* `value.strip()`: Elimina espacios al inicio y al final.
* `e.sub("[!#?]", "", value)`: Elimina los caracteres !, # y ?
* `value.title()`: Convierte la cadena a "Title Case".
* `str.title()`: Convierte la primera letra de cada palabra en mayúscula y el resto en minúscula. "hola mundo" -> "Hola Mundo".
* `str.capitalize()`: Convierte solo la primera letra de la cadena en mayúscula y el resto en minúscula. "hola mundo" -> "Hola mundo".
* Un método alternativo que puede resultar útil es crear una lista de las operaciones que se desean aplicar a un determinado conjunto de cadenas de texto.
* Un patrón más funcional como este permite modificar fácilmente el modo en que las cadenas de texto se transforman a un nivel muy alto.
* Se pueden utilizar funciones como argumentos para otras funciones, como la función integrada map, que aplica una función a una secuencia de algún tipo.
* La función `map()` en Python sirve para aplicar una función a cada elemento de un iterable (como una lista, tupla, etc.), y devolver un nuevo iterable con los resultados. `map(función, iterable)`.

```python
import re

def clean_strings(strings):
    result = []
    for value in strings:
        value = value.strip()
        value = re.sub("[!#?]", "", value)
        value = value.title()
        result.append(value)
    return result

def remove_punctuation(value):
    return re.sub("[!#?]", "", value)

# Supongamos que queremos limpiar esta lista.
states = [" Alabama ", "Georgia!", "Georgia", "georgia", "FlOrIda", " south carolina##", "West virginia?"]
res = clean_strings(states)
print(res)

# Definimos nuestra regla
clean_ops = [str.strip, remove_punctuation, str.title]
def clean_strings(strings, ops):
    result = []
    for value in strings:
        # Aplica las 3 reglas sobre la el string
        for func in ops:
            value = func(value)
        result.append(value)
    return result

res1 = clean_strings(states, clean_ops)
print(res1)

# Usando map
for x in map(remove_punctuation, states):
    print(x)
```

### Funciones anónimas (lambda)

* Python soporta las denominadas funciones anónimas o `lambda`, una forma de escribir funciones que consisten en una única sentencia, el resultado de la cual es el valor devuelto.
* Con frecuencia suele ser más rápido (y claro) pasar una función lambda, a diferencia de escribir una declaración de función completa o incluso asignar la función lambda a una variable local.
* `lambda x: len(set(x))`: lo que se hace es que toma al string, lo convierte a un conjunto, es decir, aquí no habría repeticiones, por lo que el tamaño del conjutno nos daría la cantidad de letras diferentes.

```python
# Tradicional
def short_function(x):
    return x * 2

# Lambda
equiv_anon = lambda x: x * 2

# Función que usa lambda
def apply_to_list(some_list, f):
    # esto es una función expresión, donde a cada elemento de la lista le aplicamos la función.
    return [f(x) for x in some_list]

ints = [4, 0, 1, 5, 6]
#apply_to_list(ints, lambda x: x * 2)
print(apply_to_list(ints, equiv_anon))

"""
supongamos que queremos ordenar una colección de cadenas de texto 
por el número de las letras de que se compone cada cadena.
"""
strings = ["foo", "card", "bar", "aaaa", "abab"]
strings.sort(key=lambda x: len(set(x)))
print(strings)
```
