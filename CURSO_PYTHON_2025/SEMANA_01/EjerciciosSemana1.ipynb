### 1. Calculadora de Propinas 

**Descripción**: 

*Crea un programa que calcule propinas basado en el total de la cuenta y el 
porcentaje que el usuario desea dejar.* 

El programa debe: 
* Pedir el monto total de la cuenta
* Preguntar qué porcentaje de propina desea dejar (15%, 18%, 20% u otro)
* Calcular el monto de la propina y el total a pagar
* Ofrecer la opción de dividir la cuenta entre varias personas
* Mostrar resultados 

```python
monto_total = 0.0
propina = 0.0
personas= 0
dividir_cuenta=False

while monto_total<=0:
    monto_total = float(input("\nIngrese el monto total a pagar:"))
    if(monto_total<=0):
        print("Por favor ingrese un valor no negativo y diferente de cero")

while propina<=0:
    propina =float(input("\nIngrese el porcentaje de propina a dejar:"))
    if(propina<=0):
        print("Por favor ingrese un valor no negativo y diferente de cero")

dividir_cuenta = int(input("\n¿Quiere dividir la cuenta?\n0) No\n1) Sí\n"))
dividir_cuenta = True if dividir_cuenta == 1 else False
if dividir_cuenta:
    while personas<=0:
        personas =float(input("\nIngrese entre cuantas personas quiere dividir la cuenta:"))
        if(personas<=0):
            print("Por favor ingrese un valor no negativo y diferente de cero")

monto_propina = monto_total * (propina/100)
monto_final = monto_total + monto_propina

print(f"Monto Original a Pagar: ${monto_total}")
print(f"Se seleccionó un {propina}% de propina, por lo que el monto final será {monto_final}")
if dividir_cuenta:
    monto_persona = monto_final/personas
    print(f"Se seleccionó dividir la cuenta entre {personas} personas, por lo que el monto por persona será {monto_persona}")
```

### 4. Juego: "Pares o Nones" con Dados

Descripción: 
Un juego clásico donde dos jugadores apuestan si la suma de dos dados será par o 
impar. Gana quien acierte el resultado. 
Cómo funciona: 
1. Cada jugador elige "par" o "impar" 
2. Se lanzan dos dados virtuales 
3. Se suma el resultado de los dados 
4. Gana el jugador que haya elegido correctamente si la suma es par o impar

```python
import random

hayGanador = False
while not hayGanador:
    print("\nCARRRERA DE DADOS!")
    j1 = int(input("\n[JUGADOR1]\n1)PAR\n2)IMPAR\n>"))
    j2 = int(input("\n[JUGADOR2]\n1)PAR\n2)IMPAR\n>"))
    dado1 = random.randint(1,6)
    dado2 = random.randint(1,6)
    suma = dado1+dado2
    print(f"Dado1: {dado1}, Dado2: {dado2} --> Suma {suma}")
    esPar = True if suma%2 == 0 else False
    if esPar:
        if j1==1 and j2==1 :
            print("\nAMBOS JUGADORES HAN ACERTADO!")
            hayGanador = True
            break
        if j1==1:
            print("\nEL JUGADOR 1 HA ACERTADO!")
            hayGanador = True
            break
        if j2==1:
            print("\nEL JUGADOR 2 HA ACERTADO!")
            hayGanador = True
            break
        print("\nNINGUN JUGADOR ACERTÓ!")
    else:
        if j1==2 and j2==2:
            print("\nAMBOS JUGADORES HAN ACERTADO!")
            hayGanador = True
            break
        if j1==2:
            print("\nEL JUGADOR 1 HA ACERTADO!")
            hayGanador = True
            break
        if j2==2:
            print("\nEL JUGADOR 2 HA ACERTADO!")
            hayGanador = True
            break
        print("\nNINGUN JUGADOR ACERTÓ!")
```
