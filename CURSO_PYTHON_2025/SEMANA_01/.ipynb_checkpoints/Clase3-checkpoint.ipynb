## 1. La función input()

La función `input()` nos permite recibir datos del usuario mediante el teclado.

```python
# Ejemplo básico de input()
nombre = input("Por favor, ingresa tu nombre: ")
print(f"¡Hola, {nombre}!")
```

### Características importantes de input()
- Siempre devuelve una cadena (texto)
- Podemos convertir el resultado a otros tipos de datos
- Podemos mostrar un mensaje al usuario como parámetro

```python
# Convertir el resultado de input()
edad = int(input("¿Cuántos años tienes? "))
print(f"El año que viene tendrás {edad + 1} años")

# Otro ejemplo con decimales
altura = float(input("¿Cuál es tu altura en metros? "))
print(f"Tu altura es {altura} metros")

eres_humano = bool(input("¿Eres un humano? "))
print(f"Eres Humano? {eres_humano}")
```

### Ejercicio de ejemplo: Calculadora simple

Vamos a crear una calculadora que pida dos números y realice una operación básica.

```python
# Calculadora simple
print("Calculadora simple")
num1 = float(input("Ingresa el primer número: "))
num2 = float(input("Ingresa el segundo número: "))
operacion = input("¿Qué operación deseas? (+, -, *, /): ")

if operacion == "+":
    resultado = num1 + num2
elif operacion == "-":
    resultado = num1 - num2
elif operacion == "*":
    resultado = num1 * num2
elif operacion == "/":
    if num2 != 0:
        resultado = num1 / num2
    else:
        resultado = "Error: división por cero"
else:
    resultado = "Operación no válida"

print(f"Resultado: {resultado}")
```

## 2. Bucles con while

El bucle `while` repite un bloque de código mientras se cumpla una condición.

```python
# Ejemplo básico de while
contador = 0
while contador < 5:
    print(f"Contador: {contador}")
    contador += 1  # ¡Importante! Actualizar la variable de condición

print("Bucle terminado")
```

```python
# Ejemplo de validación con while
numero = 0
while numero <= 0:
    numero = int(input("Ingresa un número positivo: "))
    if numero <= 0:
        print("¡Error! Debe ser un número positivo")

print(f"Has ingresado: {numero}")
```

**Importante: Evitar bucles infinitos**

Siempre debemos asegurarnos de que la condición del while eventualmente sea falsa.

### Juego ¡Adivina el número!

```python
# Programa sencillo: Adivina el número
import random

print("🎯 ADIVINA EL NÚMERO (1-10) 🎯")
numero_secreto = random.randint(1, 10)
intentos = 0
adivinado = False

while (not adivinado):
    intento = int(input("¿Qué número crees que es? (1-10): "))
    intentos += 1

    if (intento < 1) or (intento > 10):
        print("Por favor, ingresa un número entre 1 y 10.")
    elif intento < numero_secreto:
        print("¡Más alto! ⬆️")
    elif intento > numero_secreto:
        print("¡Más bajo! ⬇️")
    else:
        adivinado = True
        print(f"🎉 ¡Correcto! Adivinaste en {intentos} intentos.")

print("¡Juego terminado!")
```

### Píramide

```python
#Piramide de asteriscos
n = int(input("Ingresa el número de niveles para la pirámide: "))
contador = 1
print(f"\nPirámide con {n} niveles:")
while contador < n + 1:
    espacios = " " * (n - contador)      # Espacios a la izquierda
    asteriscos = "*" * (2 * contador - 1)  # Asteriscos (fórmula: 2*i - 1)
    print(espacios + asteriscos)
    contador += 1
```

### Sistema de login

```python
# Definimos las credenciales correctas como constantes
USUARIO_CORRECTO = "admin"
CONTRASEÑA_CORRECTA = "1234"

# Configuramos el número máximo de intentos permitidos
intentos_maximos = 3

# Inicializamos el contador de intentos en 0
intentos = 0

# Variable booleana para controlar si se concedió acceso
acceso_concedido = False

# Mostramos el título del sistema
print("=== SISTEMA DE LOGIN ===")

# Bucle while: se ejecuta mientras haya intentos disponibles y no se haya concedido acceso
while (intentos < intentos_maximos) and (not acceso_concedido):

    # Solicitamos al usuario que ingrese sus credenciales
    usuario = input("Usuario: ")
    contraseña = input("Contraseña: ")

    # Verificamos si las credenciales son correctas
    if (usuario == USUARIO_CORRECTO) and (contraseña == CONTRASEÑA_CORRECTA):

        # Si son correctas, mostramos mensaje de éxito
        print("¡Acceso concedido! Bienvenido al sistema.")

        # Cambiamos el estado de acceso a True
        acceso_concedido = True
    else:
        # Si son incorrectas, incrementamos el contador de intentos
        intentos += 1

        # Calculamos cuántos intentos le quedan al usuario
        intentos_restantes = intentos_maximos - intentos

        # Verificamos si todavía quedan intentos
        if intentos_restantes > 0:
            # Informamos al usuario sobre intentos restantes
            print(f"Credenciales incorrectas. Te quedan {intentos_restantes} intentos.")
        else:
            # Si no quedan intentos, mostramos mensaje de bloqueo
            print("Has agotado todos tus intentos. Sistema bloqueado.")

# Fuera del bucle, mostramos un mensaje final según si se concedió acceso o no
if acceso_concedido:
    print("Puedes comenzar a usar el sistema.")
else:
    print("Por favor, contacta al administrador para recuperar el acceso.")
```
