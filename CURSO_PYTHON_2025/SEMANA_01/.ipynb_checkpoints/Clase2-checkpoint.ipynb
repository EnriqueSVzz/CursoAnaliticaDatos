## Operadores en Python

Antes de aprender condicionales, repasemos los operadores que nos permiten crear condiciones y realizar cálculos.

### Operadores Aritméticos

Realizan operaciones matemáticas básicas:

| Operador | Operación      | Ejemplo   | Resultado |
|----------|----------------|-----------|-----------|
| `+`      | Suma           | `5 + 3`   | `8`       |
| `-`      | Resta          | `10 - 4`  | `6`       |
| `*`      | Multiplicación | `6 * 7`   | `42`      |
| `/`      | División       | `15 / 2`  | `7.5`     |
| `//`     | División entera| `15 // 2` | `7`       |
| `%`      | Módulo (resto) | `15 % 2`  | `1`       |
| `**`     | Potencia       | `2 ** 3`  | `8`       |

````python
# Practiquemos con operadores aritméticos
a = 20
b = 50

print("a =", a, ", b =", b)
print("Suma:", a + b)
print("Resta:", a - b)
print("Multiplicación:", a * b)
print("División:", a / b)
print("División entera:", a // b)
print("Módulo (resto):", a % b)
print("Potencia:", a ** b)

# Ejemplo práctico
precio_producto = 77
cantidad = 2
total = precio_producto * cantidad
print(f"\nPrecio total por {cantidad} productos: ${total}")
````

### Operadores de Comparación

Comparan dos valores y devuelven `True` o `False`:

| Operador | Significado       | Ejemplo   | Resultado |
|----------|-------------------|-----------|-----------|
| `==`     | Igual a           | `5 == 5`  | `True`    |
| `!=`     | Diferente de      | `5 != 3`  | `True`    |
| `>`      | Mayor que         | `10 > 5`  | `True`    |
| `<`      | Menor que         | `10 < 5`  | `False`   |
| `>=`     | Mayor o igual que | `10 >= 10`| `True`    |
| `<=`     | Menor o igual que | `10 <= 5` | `False`   |

````python
# Veamos cómo funcionan los operadores de comparación
x = 30
y = 5.5

print("x =", x, ", y =", y)
print("x == y:", x == y) #Falso
print("x != y:", x != y) #Verdadero
print("x > y:", x > y) #Verdadero
print("x < y:", x < y) #Falso
print("x >= y:", x >= y) #Verdadero
print("x <= y:", x <= y) #Falso

# Comparando strings
nombre1 = "Ana"
nombre2 = "ana"
print(f"\n'{nombre1}' == '{nombre2}':", nombre1 == nombre2)
````

### Operadores Lógicos

Combinan condiciones y devuelven `True` o `False`:

| Operador | Descripción                          | Ejemplo           |
|----------|--------------------------------------|-------------------|
| `and`    | True si AMBAS condiciones son True   | `True and True`   |
| `or`     | True si AL MENOS UNA condición es True | `True or False`   |
| `not`    | Invierte el resultado (NOT)          | `not True`        |

````python
# Ejemplos de operadores lógicos
edad = 30
tiene_licencia = False
tiene_auto = True

print("Edad:", edad, "| Licencia:", tiene_licencia, "| Auto:", tiene_auto)

# AND: ambas deben ser True
puede_conducir = edad >= 18 and tiene_licencia
print("Puede conducir?", puede_conducir)

# OR: al menos una debe be True
puede_viajar = tiene_auto or tiene_licencia
print("Puede viajar?", puede_viajar)

# NOT: invierte el valor
es_menor = not (edad >= 18)
print("Es menor de edad?", es_menor)

# Combinaciones complejas
condicion_compleja = (edad > 20) and (tiene_auto or tiene_licencia)
print("Condición compleja:", condicion_compleja)
````

### Tabla de Verdad de Operadores Lógicos

#### Operador AND (`and`)
| A     | B     | A and B |
|-------|-------|---------|
| True  | True  | True    |
| True  | False | False   |
| False | True  | False   |
| False | False | False   |

**Regla:** Solo es `True` cuando AMBAS condiciones son `True`.

---

#### Operador OR (`or`)
| A     | B     | A or B  |
|-------|-------|---------|
| True  | True  | True    |
| True  | False | True    |
| False | True  | True    |
| False | False | False   |

**Regla:** Es `True` cuando AL MENOS UNA condición es `True`.

---

#### Operador NOT (`not`)
| A     | not A |
|-------|-------|
| True  | False |
| False | True  |

**Regla:** Invierte el valor booleano.

---

#### Combinación Compleja: `A and (B or C)`
| A     | B     | C     | B or C | A and (B or C) |
|-------|-------|-------|--------|----------------|
| True  | True  | True  | True   | True           |
| True  | True  | False | True   | True           |
| True  | False | True  | True   | True           |
| True  | False | False | False  | False          |
| False | True  | True  | True   | False          |
| False | True  | False | True   | False          |
| False | False | True  | True   | False          |
| False | False | False | False  | False          |

## ¿Qué son las estructuras condicionales?

Son instrucciones que permiten que nuestro programa tome decisiones y ejecute diferentes bloques de código dependiendo de si una condición es Verdadera (True) o Falsa (False).

**Ejemplo de la vida real:**
- **SI** está lloviendo → llevar paraguas
- **SI NO** → no llevar paraguas

````python
# Sintaxis básica del if
edad = 20

if edad >= 18:
    print("¡Eres mayor de edad!")
    print("Puedes votar en las elecciones")

print("Esta línea siempre se ejecuta")
````

### La sentencia `if`

**Sintaxis:**
```python
if condición:
    # código que se ejecuta si la condición es True
    # ¡importante la indentación (tabulación)!

**Reglas importantes:**
```
- La condición debe evaluar a `True` o `False`
- Los dos puntos (:) al final de la condición son obligatorios
- La indentación (espacio o tab) define el bloque de código

````python
# Ejemplo if-elif-else
calificacion = 78.9

if calificacion >= 90:
    print("¡Excelente!")
elif calificacion >= 80:
    print("¡Muy bien!")
elif calificacion >= 70:
    print("Bien")
elif calificacion >= 60:
    print("Necesitas mejorar")
else :
    print("Reprobado :(")

print("Evaluación completada")
````

````python
# Operadores lógicos: and, or, not

# AND: ambas condiciones deben ser True
edad = 40
tiene_licencia = True

if edad >= 18 and tiene_licencia:
    print("Puedes conducir legalmente")
else:
  print("No puedes conducir legalmente")

# OR: al menos una condición debe ser True
es_fin_de_semana = False
es_feriado = False

if es_fin_de_semana or es_feriado:
    print("¡Día de descanso!")
else:
  print("Debes trabajar")

# NOT: invierte el valor booleano
llueve = True

if not llueve:
    print("Hace buen día para pasear")
else:
  print("LLeva paraguas")
````

````python
# Condicionales dentro de condicionales (condicionales aninadas)
edad = 19
tiene_entrada = False
es_mayor_de_edad = edad >= 18

if es_mayor_de_edad:
    print("Eres mayor de edad")
    if tiene_entrada:
        print("Puedes entrar al concierto")
    else:
        print("Necesitas comprar entrada primero")
else:
    print("Eres menor de edad, no puedes entrar")
````

### Ejercicio 2: Calculadora de descuentos

Crea un programa que:
- Declare 5 variables y asigneles un valor de precio
-Declare una variable llamada compra que sea la suma de todos los productos
- Aplique los descuentos:
  - Si compra >= 100, aplicar 10 porciento de descuento
  - Si compra >= 200, aplicar 20 porciento de descuento
  - Si compra >= 500, aplicar 30 porciento de descuento
- Muestre el monto final a pagar

````python
# Ejercicio 2: Calculadora de descuentos

compra1=120
compra2=30
compra3=134
compra4=68.6
compra5=113.2

total = compra1+compra2+compra3+compra4+compra5

if total >=100 and total <200:
    print(f"Monto Original: {total}. Se aplica un 10% de descuento\nMonto Final: {total*0.90}")
elif total >=200 and total <500:
    print(f"Monto Original: {total}. Se aplica un 20% de descuento\nMonto Final: {total*0.80}")
elif total >=500:
    print(f"Monto Original: {total}. Se aplica un 30% de descuento\nMonto Final: {total*0.30}")
````
