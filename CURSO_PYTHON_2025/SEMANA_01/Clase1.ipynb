## ¿Qué es Python?

Python es un lenguaje de programación:
- **Interpretado**: No necesita compilación, se ejecuta línea por línea
- **De alto nivel**: Fácil de leer y escribir
- **Multipropósito**: Se usa en web, datos, inteligencia artificial, automatización, etc.

**¿Por qué aprender Python?**
- Es uno de los lenguajes más populares del mundo
- Sintaxis clara y fácil de aprender
- Gran comunidad y muchas librerías disponibles



## ¿Cómo "hablan" las computadoras?

Las computadoras entienden **código binario (0 y 1)**, pero los humanos escribimos en lenguajes de programación. Existen diferentes tipos:

### Lenguajes Interpretados (como Python)

**Flujo de trabajo:**
Código → Intérprete → Ejecución línea por línea

**Características:**
- Más flexibilidad
- Ejecución inmediata
- Menor eficiencia
- Mensajes de error en tiempo real

**Ejemplos:** Python, JavaScript, Ruby

### Lenguajes Compilados (como C++)

**Flujo de trabajo:**
Código → Compilador → Ejecutable → Ejecución

**Características:**
- Mayor velocidad
- Optimización avanzada
- Menos flexibilidad
- Detección de errores antes de ejecutar

**Ejemplos:** C, C++, Rust, Go

### Visualicemos la diferencia con un ejemplo simple
**LENGUAJE INTERPRETADO (Python)**
1. Escribes el código
2. El intérprete lee línea por línea
3. Se ejecuta inmediatamente
4. Si hay error, se detiene y muestra dónde

**LENGUAJE COMPILADO (C)**
1. Escribes el código
2. El compilador convierte TODO a máquina
3. Si hay errores, NO se crea ejecutable
4. Si compila, se ejecuta el programa completo

### Lenguajes de Alto Nivel vs Bajo Nivel

**Lenguajes de BAJO nivel (ej: Ensamblador)**
- Cercanos al hardware de la computadora
- Difíciles de leer y escribir para humanos
- Máximo control sobre el hardware
- Poco portables entre diferentes computadoras

**Lenguajes de MEDIO nivel (ej: C)**
- Balance entre control y facilidad de uso
- Más comprensibles para humanos
- Buen control del hardware

**Lenguajes de ALTO nivel (ej: Python)**
- Muy fáciles de leer y escribir
- Alejados del hardware
- Menos control sobre detalles específicos
- Muy portables entre diferentes sistemas

### Ejemplo que muestra la diferencia de abstracción

**Mismo concepto en diferentes niveles**

**ALTO NIVEL (Python)**
- suma = 5 + 3

**MEDIO NIVEL (C):**
- int suma = 5 + 3;

**BAJO NIVEL (Ensamblador x86):**
1. mov eax, 5
2. add eax, 3
3. mov [suma], eax

### ¿Dónde está Python en este espectro?

**Python es:**
- **Lenguaje interpretado**: Se ejecuta línea por línea
- **Lenguaje de alto nivel**: Muy fácil de leer y escribir
- **Multiplataforma**: Funciona en Windows, Mac, Linux
- **Dinámico**: No necesitas declarar tipos de variables

**Ventajas para principiantes:**
- Sintaxis clara y legible
- Feedback inmediato al ejecutar
- No necesita compilación previa
- Gran comunidad y recursos

### Variables: contenedores para almacenar datos

Empezemos por lo más básico, las variables. Las variables son el nombre que le damos una propiedad con la cual vamos a trabajar.

```python
# Creando variables
nombre = 'Enrique'
edad = 23
altura = 1.77
es_estudiante = False
es_trabajador = True

# Mostrando las variables
print("Nombre:", nombre)
print("Edad:", edad)
print("Altura:", altura)
print("Es estudiante:", es_estudiante)
print("\n")
print("hola que tal, mi nombre es", nombre, ". Tengo", edad, "años. Y si se preguntan, ¿Soy estudiante? pues esto es:",es_estudiante)
```

```python
presetnacion = "Hola, mi nombre es "
nombre = "Jorge"
suma = presetnacion + nombre
print(suma)
```

### 2. Investigar: ¿qué otros tipos de datos existen en Python?

* Actualmente existen estos tipos de datos en python. Algunos son considerados primitivos o básicmos como lo serían el `int`, `float`, `str` o el `bool`. Pero también existen algunos que llaman "compuestos" o "complejos", es decir que a partir de estos podemos construir nuevos tipos de estructuras.
* Estos tipos de datos podemos verlos en la tabla, por ejemplo: los diccionarios, las listas, tuplas o los conjuntos set.

| Tipo en Python                     | Ejemplos                           |
| ---------------------------------- | ---------------------------------- |
| `int` (enteros)                    | `6`, `5`, `-2`, `675`              |
| `float` (punto flotante)           | `4.5`, `6.43`, `9.7e10`            |
| `complex` (números complejos)      | `1j+45`, `5j-87`                   |
| `str` (cadenas de texto)           | `'Pepe'`, `'Juan'`, `'Coche'`      |
| `bool` (booleanos)                 | `True`, `False` (también `None`)   |
| `dict` (diccionarios)              | `{color='red'}`, `dict(num=4)`     |
| `list` (listas)                    | `[1, 42, 3j, 'Hello']`             |
| `tuple` (tuplas)                   | `(3, 'Coche', 8.9)`                |
| `set` (conjuntos mutables)         | `{3, 4, 5, 'Juan', 3.21}`          |
| `frozenset` (conjuntos inmutables) | `frozenset([1, 2, 'Pepe', 5j])`    |
| `range` (rangos)                   | `range(5, 10)`, `range(-5, 20, 4)` |
| `bytes` (binarios)                 | `b'3y4'`, `bytes(45)`              |
