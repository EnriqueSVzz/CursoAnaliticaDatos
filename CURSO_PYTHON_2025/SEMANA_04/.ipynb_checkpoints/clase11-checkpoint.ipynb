# Clase 11: Introducción a pandas

```python
# Primero, importamos pandas (por convención se usa el alias pd)
import pandas as pd
import numpy as np  # También importamos numpy que suele usarse junto con pandas
```

# Data frame

```python
# Creación de un DataFrame desde un arreglo
data = {
    'Nombre': ['Juan', 'María', 'Carlos', 'Ana'],
    'Edad': [25, 32, 28, 35],
    'Ciudad': ['Madrid', 'Barcelona', 'Valencia', 'Sevilla']
}

### Crear el dataframe apartir de un diccionario
df = pd.DataFrame(data)
df
```

* Los dataframes están copuestos de series, que podría entenderse como una columna que forma parte del excel.

```python
# Creación de una Serie
edades = pd.Series([25, 32, 28, 35], name='Edad')
edades
```

## Lectura y escritura de datos

* `df.to_csv` guardar los datos en un csv.

```python
# Lectura de un archivo CSV (comentar para el ejemplo, ya que no existe el archivo)
# df = pd.read_csv('datos.csv')

# Escritura a CSV
df.to_csv('datos_ejemplo2.csv', index=False)

# Lectura desde Excel
# df = pd.read_excel('datos.xlsx')

# Lectura desde una URL
url = 'https://raw.githubusercontent.com/mwaskom/seaborn-data/master/iris.csv'
iris = pd.read_csv(url)
iris.head()  # Muestra las primeras 5 filas
```

## Exporación básica de datos

* La función `display()` se usa para mostrar objetos de forma más rica e interactiva en entornos como `Jupyter Notebook`, `Google Colab` o cualquier entorno basado en `IPython`.
* valores unicos --> `display(iris['species'].unique())`

```python
from IPython.display import display

# Información básica del DataFrame
print("* Información del DataFrame:")
display(iris.info())

# Estadísticas descriptivas
print("\n* Estadísticas descriptivas:")
display(iris.describe())

# Primeras filas
print("\n* Primeras 3 filas:")
display(iris.head(3))

# Últimas filas
print("\n* Últimas 2 filas:")
display(iris.tail(2))

# Forma del DataFrame (filas, columnas)
print("\n* Forma del DataFrame:")
display(iris.shape)

# Nombres de columnas
print("\n* Nombres de columnas:")
display(iris.columns)

# Valores únicos en una columna
print("\n* Especies únicas:")
display(iris['species'].unique())
```

## Seleccion y filtrado de datos

```python
### Selección de una columna (devuelve Series)
display(iris['sepal_length'])

### Selección de múltiples columnas
display(iris[['sepal_length', 'species']])
```

```python
### Selección por posición (iloc)
print("Primeras 3 filas, primeras 2 columnas:")
display(iris.iloc[:3, :2])

### Selección por etiqueta (loc), del segundo registro al 3 registro sobre sepal y species
print("\nFiltrado por índice y columnas:")
display(iris.loc[2:5, ['sepal_length', 'species']])
```

* Esta muy interesante esto `iris.query`

```python
### Filtrado con condiciones
print("\nPlantas con sépalo mayor a 5 cm:")
display(iris[iris['sepal_length'] > 5])

# Filtrado con múltiples condiciones
print("\nPlantas con sépalo >5 y pétalo <1.5:")
display(iris[(iris['sepal_length'] > 5) & (iris['petal_length'] < 1.5)])

# Query (filtrado con expresión de cadena)
print("\nUsando query:")
display(iris.query('sepal_length > 5 and petal_length < 1.5'))
```

## Manipulación de datos

```python
### Añadir nueva columna
iris['sepal_area'] = iris['sepal_length'] * iris['sepal_width']
display(iris.head())

### Renombrar columnas
iris_renombrado = iris.rename(columns={
    'sepal_length': 'longitud_sepalo',
    'sepal_width': 'ancho_sepalo'
})
display(iris_renombrado.head(2))

### Eliminar columnas
iris_sin_area = iris.drop(columns=['sepal_area'])
display(iris_sin_area.head(2))

### Ordenar valores
print("Ordenado por longitud de sépalo descendente:")
display(iris.sort_values('sepal_length', ascending=False).head())

### Valores faltantes (aunque en este dataset no hay)
print("\nValores nulos por columna:")
display(iris.isnull().sum())
```

## Operaciones agrupadas

```python
# GroupBy - Agrupación por especie
grupo_especies = iris.groupby('species')

# Estadísticas por grupo
print("Media por especie:")
display(grupo_especies.mean())

# Contar elementos por grupo
print("\nConteo por especie:")
display(grupo_especies.size())

# Aplicar múltiples funciones de agregación
print("\nMúltiples estadísticas:")
display(grupo_especies.agg(['mean', 'std', 'min', 'max']))
```

## Combinación de data frames

```python
# Crear otro DataFrame para combinar
data2 = {
    'species': ['setosa', 'versicolor', 'virginica'],
    'color': ['blanco', 'multicolor', 'púrpura']
}
df_colores = pd.DataFrame(data2)

# Merge
print("Combinación por especie:")
pd.merge(iris, df_colores, on='species')
```

## Operaciones con series

```python
# Operaciones vectorizadas
print("Suma de longitudes:")
display(iris['sepal_length'] + iris['petal_length'])

# Aplicar funciones
print("\nLongitud de sépalo redondeada:")
display(iris['sepal_length'].apply(lambda x: round(x, 1)))

# Mapear valores
print("\nMapeo de especies a códigos:")
display(iris['species'].map({'setosa': 1, 'versicolor': 2, 'virginica': 3}).head(150))
```
