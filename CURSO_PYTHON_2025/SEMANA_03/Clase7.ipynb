## 1. Introducción al Manejo de Archivos

Los archivos nos permiten **persistir datos** beyond the execution of a program. Python ofrece funciones built-in para trabajar con diferentes tipos de archivos.

**¿Por qué es importante?**
- Guardar datos permanentemente
- Procesar información de archivos existentes
- Intercambiar datos entre programas
- Generar reportes y logs

````python
# Ejemplo básico: Crear y leer un archivo
print("=== MANEJO BÁSICO DE ARCHIVOS ===")

# Escribir en un archivo
with open("ejemplo.txt", "w") as archivo:
    archivo.write("Hola, mundo!\n")
    archivo.write("Este es un archivo de ejemplo.\n")
    archivo.write("Python es genial!\n")
    archivo.write("Ejemplo de texto!\nHola")

print("Archivo 'ejemplo.txt' creado exitosamente!")
````

### 2. Modos de Apertura de Archivos

Python ofrece diferentes modos para abrir archivos:

| Modo | Descripción |
|------|-------------|
| `'r'` | Lectura (`read`) - `Valor por defecto` |
| `'w'` | Escritura (`write`) - Sobrescribe el archivo |
| `'a'` | Append - `Agrega al final` del archivo |
| `'x'` | Creación - `Fallo si el archivo existe` |
| `'b'` | Modo `binario` |
| `'t'` | Modo texto (default) |
| `'+'` | `Lectura y escritura` |

**Ejemplos:**
- `'r'` - Solo lectura
- `'w+'` - `Lectura y escritura (sobrescribe)`
- `'a+'` - `Lectura y escritura (append)`

````python
# Diferentes modos de apertura
print("=== MODOS DE APERTURA en GoogleDrive ===")

# Modo escritura ('w') - Sobrescribe el archivo
#with open(drive_path + "datos.txt", "w") as archivo:
with open("datos.txt", "w+") as archivo:
    archivo.write("Línea 1\n")
    archivo.write("Línea 2\n")
    archivo.write("Línea 3\n")
    archivo.write("Línea 4\n")

print("Archivo escrito en modo 'w'")

# Modo append ('a') - Agrega al final
#with open(drive_path +"datos.txt", "a") as archivo:
with open("datos.txt", "a") as archivo:
    archivo.write("Línea 3 (agregada con append)\n")
    archivo.write("Línea 4 (también agregada)\n")
    archivo.write("Enrique\n")

print("Contenido agregado en modo 'a'")

# Leer el archivo completo
#with open(drive_path +"datos.txt", "r") as archivo:
with open("datos.txt", "r") as archivo:
    contenido = archivo.read()
    print("\nContenido completo del archivo:")
    print(f"Contenido: {contenido}")

# Leer línea por línea
print("\nLeyendo línea por línea:")
#with open(drive_path +"datos.txt", "r") as archivo:
with open("datos.txt", "r") as archivo:
    for i, linea in enumerate(archivo, 1):
        print(f"Línea {i}: {linea.strip()}")
````

### 3. Trabajando con Archivos CSV

Los archivos CSV (Comma-Separated Values) son muy comunes para intercambiar datos tabulares.

***NOTAS***
* `writer = csv.writer(archivo)`: se le pase el context del archivo que quiere trabajar.

````python
# Trabajando con archivos CSV
print("=== ARCHIVOS CSV ===")

import csv

# Escribir un archivo CSV
datos = [
    ["Nombre", "Edad", "Ciudad"],
    ["Ana", 25, "Madrid"],
    ["Carlos", 30, "Barcelona"],
    ["María", 28, "Valencia"]
]

### Creando nuestro archivo CSV.
#with open(drive_path + "personas.csv", "w", newline='') as archivo:
with open("personas.csv", "w", newline='') as archivo:
    writer = csv.writer(archivo)
    writer.writerows(datos)

print("Archivo CSV creado exitosamente!")

# Leer el archivo CSV
print("\nLeyendo archivo CSV:")
#with open(drive_path + "personas.csv", "r") as archivo:
with open("personas.csv", "r") as archivo:
    reader = csv.reader(archivo)
    for fila in reader:
        print(fila)

# Leer como diccionario (más conveniente)
print("\nLeyendo CSV como diccionario:")
#with open(drive_path + "personas.csv", "r") as archivo:
with open("personas.csv", "r") as archivo:
    reader = csv.DictReader(archivo)
    for fila in reader:
        print(f"{fila['Nombre']} - {fila['Edad']} años - {fila['Ciudad']}")
````

### 4. Introducción a las Excepciones

Las excepciones son errores que ocurren durante la ejecución de un programa. En lugar de dejar que el programa se caiga, podemos **manejarlas** gracefuly.

**Tipos comunes de excepciones:**
- `FileNotFoundError`: Archivo no encontrado
- `ValueError`: Valor incorrecto
- `TypeError`: Tipo de dato incorrecto
- `IndexError`: Índice fuera de rango
- `KeyError`: Clave no encontrada en diccionario
- `ZeroDivisionError`: División por cero

````python
# Manejo básico de excepciones
print("=== MANEJO DE EXCEPCIONES ===")

# Sin manejo de excepciones (esto causará error)
try:
    resultado = 10 / 0
except ZeroDivisionError:
    print("¡Error! No se puede dividir por cero.")

# Múltiples excepciones
try:
    numero = int("no_es_un_numero")
    resultado = 10 / 0
except ValueError:
    print("Error: Valor no convertible a número")
except ZeroDivisionError:
    print("Error: División por cero")

# Excepción general
try:
    with open("archivo_inexistente.txt", "r") as archivo:
        contenido = archivo.read()
except Exception as e:
    print(f"Ocurrió un error: {type(e).__name__} - {e}")
````

### 5. Try-Except-Finally

La estructura completa `try-except-finally` nos permite:
- `try`: Intentar ejecutar código que podría fallar
- `except`: Manejar las excepciones
- `finally`: ***Ejecutar código que siempre debe correr (limpieza)***

````python
# Try-Except-Finally completo
print("=== TRY-EXCEPT-FINALLY ===")
try:
    num1 = float(input("Primer número: "))
    num2 = float(input("Segundo número: "))
    resultado = num1 / num2
    print(f"Resultado: {resultado}")

except ValueError:
    print("Error: Debes ingresar números válidos")

except ZeroDivisionError:
    print("Error: No se puede dividir por cero")

finally:
    print("Esta línea siempre se ejecuta, haya error o no")
````

### 6. Excepciones Comunes con Archivos

Al trabajar con archivos, estas son algunas excepciones comunes:

````python
# Excepciones comunes con archivos
print("=== EXCEPCIONES COMUNES CON ARCHIVOS ===")

# FileNotFoundError
try:
    #with open(drive_path+"archivo_que_no_existe.txt", "r") as archivo:
    with open("archivo_que_no_existe.txt", "r") as archivo:
        contenido = archivo.read()
except FileNotFoundError:
    print("Error: El archivo no existe")
except PermissionError:
    print("Error: No tienes permisos para leer el archivo")
except IOError:
    print("Error de entrada/salida")

# Manejo elegante de archivos
#nombre_archivo = drive_path + "datos_importantes.txt"
nombre_archivo = "datos_importantes.txt"

try:
    with open(nombre_archivo, "r") as archivo:
        print("Contenido del archivo:")
        print(archivo.read())

except FileNotFoundError:
    print(f"El archivo '{nombre_archivo}' no existe. Creándolo...")
    with open(nombre_archivo, "w") as archivo:
        archivo.write("Este es un archivo nuevo\n")
    print("Archivo creado exitosamente!")

except Exception as e:
    print(f"Error inesperado: {e}")
````
