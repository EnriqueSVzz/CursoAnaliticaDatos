## Programa 1: Gestor de tareas

```python
def gestor_tareas():


    ARCHIVO = "tareas.txt"

    def cargar_tareas():
        try:
            #with open(dir+ARCHIVO, 'r') as archivo:
            ### Leemosel archivo y regresemos todas las lineas del archivo con trim.
            with open(ARCHIVO, 'r') as archivo:
                return [linea.strip() for linea in archivo.readlines()]
        except FileNotFoundError:
            return []

    def guardar_tareas(tareas):
        #with open(dir+ARCHIVO, 'w') as archivo:
        ### Guardemos la nueva tarea en nuestra base de datos.
        with open(ARCHIVO, 'w') as archivo:
            for tarea in tareas:
                archivo.write(tarea + '\n')

    def mostrar_tareas(tareas):
        ### Mostrar el contenido de tareas cargadas en el arreglo.
        if not tareas:
            print("No hay tareas pendientes.")
        else:
            print("\n--- TAREAS PENDIENTES ---")
            for i, tarea in enumerate(tareas, 1):
                print(f"{i}. {tarea}")

    ### Cargamos todas las tareas que se han registrado.
    tareas = cargar_tareas()

    while True:
        print("\n--- GESTOR DE TAREAS ---")
        print("1. Agregar tarea")
        print("2. Marcar como completada")
        print("3. Mostrar tareas")
        print("4. Salir")

        try:
            opcion = int(input("Selecciona una opción: "))

            if opcion == 1:
                nueva_tarea = input("Ingresa la nueva tarea: ")
                tareas.append(nueva_tarea)
                guardar_tareas(tareas)
                print("Tarea agregada correctamente.")

            elif opcion == 2:
                mostrar_tareas(tareas)
                try:
                    num = int(input("Número de tarea a completar: "))
                    if 1 <= num <= len(tareas):
                        ### Eliminamos la tarea completada
                        tarea_completada = tareas.pop(num-1)
                        ### Guardamos los cambios (sobreescribiendo).
                        guardar_tareas(tareas)
                        print(f"Tarea completada: {tarea_completada}")
                    else:
                        print("Número inválido.")
                except ValueError:
                    print("Debes ingresar un número válido.")

            elif opcion == 3:
                mostrar_tareas(tareas)

            elif opcion == 4:
                print("¡Hasta luego!")
                break

            else:
                print("Opción no válida.")

        except ValueError:
            print("Por favor, ingresa un número válido.")
```

```python
### Ejecutamos nuestras tareas
gestor_tareas()
```

## Programa 2: Generador de poemas automático

```python

import random
from IPython.display import display, HTML


sustantivos = ["luna", "sombra", "río", "silencio", "viento",
               "destino", "alba", "olvido", "piel", "espejo"]
verbos = ["canta", "llora", "acaricia", "danza", "sueña",
          "despierta", "muere", "renace", "besa", "huye"]
adjetivos = ["frágil", "eterno", "azul", "roto", "transparente",
             "lejano", "dulce", "amargo", "solo", "inmóvil"]
conectores = ["mientras", "bajo", "sobre", "como", "con",
              "sin", "hacia", "entre", "antes de", "después de"]


estructuras = [
    lambda: f"El {random.choice(sustantivos)} {random.choice(verbos)}",
    lambda: f"{random.choice(adjetivos).capitalize()} {random.choice(sustantivos)}",
    lambda: f"{random.choice(conectores).capitalize()} el {random.choice(sustantivos)}",
    lambda: f"{random.choice(verbos).capitalize()} {random.choice(conectores)} la {random.choice(sustantivos)}"
]


def generar_poema(estrofas=3, versos_por_estrofa=4):
    poema = []
    for _ in range(estrofas):
        estrofia = []
        for _ in range(versos_por_estrofa):
            verso = random.choice(estructuras)()
            estrofia.append(verso)
        poema.append("  \n".join(estrofia))
    return "\n\n".join(poema)


poema = generar_poema()
print(poema)
```

## Programa 3: Adivina el numero


```python
import random

def juego_adivinanza():
    """
    Juego de adivinanza con sistema de puntuación
    """
    puntuacion = 0
    record = 0

    def cargar_record():
        try:
            #with open(dir+"record.txt", "r") as archivo:
            with open("record.txt", "r") as archivo:
                return int(archivo.read().strip())
        except (FileNotFoundError, ValueError):
            return 0

    def guardar_record(nuevo_record):
        #with open(dir+"record.txt", "w") as archivo:
        with open("record.txt", "w") as archivo:

            archivo.write(str(nuevo_record))

    record = cargar_record()

    print("🎯 JUEGO DE ADIVINANZA 🎯")
    print(f"Record actual: {record} puntos")

    while True:
        numero_secreto = random.randint(1, 100)
        intentos = 0
        max_intentos = 7

        print(f"\nNuevo juego! Adivina el número entre 1 y 100")
        print(f"Tienes {max_intentos} intentos")

        while intentos < max_intentos:
            try:
                guess = int(input(f"Intento {intentos + 1}: "))

                if guess == numero_secreto:
                    puntos = (max_intentos - intentos) * 10
                    puntuacion += puntos
                    print(f"🎉 ¡Correcto! Ganaste {puntos} puntos")
                    break
                elif guess < numero_secreto:
                    print("⬆️ Más alto")
                else:
                    print("⬇️ Más bajo")

                intentos += 1

            except ValueError:
                print("Por favor, ingresa un número válido.")

        else:
            print(f"❌ Perdiste. El número era: {numero_secreto}")

        # Actualizar record si es necesario
        if puntuacion > record:
            record = puntuacion
            guardar_record(record)
            print(f"🏆 ¡Nuevo record! {record} puntos")

        print(f"Puntuación actual: {puntuacion}")

        continuar = input("\n¿Quieres jugar otra vez? (s/n): ").lower()
        if continuar != 's':
            print(f"¡Gracias por jugar! Puntuación final: {puntuacion}")
            break
```

```python
juego_adivinanza()
```
