### 1. Simulador de Gestión de Finanzas Personales

***Funcionalidades adicionales:*** 

* Guardar automáticamente todos los registros en un archivo CSV al cerrar el programa.
* Cargar datos previos al iniciar la aplicación.
* Implementar try-except para:
  - Validar que el monto sea numérico y positivo
  - Manejar errores de archivo (permisos, archivo corrupto, etc.)
  - Verificar que las categorías existan antes de agregar transacciones 
* Exportar resumen financiero a un archivo de texto 
* Opción para hacer backup de los datos financieros 

```python
import csv

def getMenuOption():
    while True:
        try:
            print("\n=== MENU PRINCIPAL ===")
            print("1) Agregar Ingreso")
            print("2) Agregar Gasto")
            print("3) Ver Resumen")
            print("4) Salir")
            opcion = int(input("> "))
            if opcion in [1, 2, 3, 4]:
                return opcion
            else:
                print("Opción fuera de rango.")
        except Exception as e:
            print(f"Ocurrió un error: {type(e).__name__} - {e}")

def getSubMenuOption():
    while True:
        try:
            print("\n=== SUBMENÚ ===")
            print("1) Exportar Información")
            print("2) Crear Backup")
            print("3) Volver al menú principal")
            opcion = int(input("> "))
            if opcion in [1, 2, 3]:
                return opcion
            else:
                print("Opción fuera de rango.")
        except Exception as e:
            print(f"Ocurrió un error: {type(e).__name__} - {e}")

def addTransaction(transacciones, tipo):
    movimiento = {
        "tipo": tipo,
        "monto": 0.0,
        "descrip": "",
        "categ": ""
    }

    while True:
        try:
            monto = float(input("Ingrese el monto: "))
            if tipo == "I" and monto < 0:
                print("No se permiten montos negativos para ingresos.")
            elif tipo == "G" and monto > 0:
                print("No se permiten montos positivos para gastos.")
            else:
                movimiento["monto"] = monto
                break
        except Exception as e:
            print(f"Ocurrió un error: {type(e).__name__} - {e}")

    movimiento["descrip"] = input("Ingrese la descripción: ")
    movimiento["categ"] = input("Ingrese la categoría: ")
    transacciones.append(movimiento)

def showSummaryReport(transacciones):
    total = sum(t["monto"] for t in transacciones)
    gasto_categoria = {}

    for transaccion in transacciones:
        cat = transaccion["categ"]
        gasto_categoria[cat] = gasto_categoria.get(cat, 0.0) + transaccion["monto"]

    print(f"\n=== RESUMEN ===\nBalance total: ${total:.2f}")
    for categoria, monto in gasto_categoria.items():
        print(f"Categoría: [{categoria}] -> ${monto:.2f}")

def loadData(filename="backlog.csv"):
    print("=== CARGANDO INFORMACIÓN ===")
    transacciones = []
    try:
        with open(filename, "r") as archivo:
            reader = csv.DictReader(archivo)
            for fila in reader:
                fila["monto"] = float(fila["monto"])
                transacciones.append(fila)
        print("Transacciones cargadas correctamente.")
    except FileNotFoundError:
        print("El archivo no existe. Se iniciará con una lista vacía.")
    except PermissionError:
        print("No tienes permisos para leer el archivo.")
    except Exception as e:
        print(f"Ocurrió un error al cargar: {type(e).__name__} - {e}")
    return transacciones

def saveData(transacciones, filename="backlog.csv"):
    print(f"\n=== GUARDANDO DATOS EN {filename} ===")
    try:
        with open(filename, "w", newline='') as archivo:
            campos = ["tipo", "monto", "descrip", "categ"]
            writer = csv.DictWriter(archivo, fieldnames=campos)
            writer.writeheader()
            writer.writerows(transacciones)
        print("Datos guardados correctamente.")
    except Exception as e:
        print(f"Ocurrió un error al guardar: {type(e).__name__} - {e}")

def exportData(transacciones, filename="export.txt", encoding="utf-8"):
    print(f"\n=== EXPORTANDO A {filename} ===")
    try:
        with open(filename, "w", newline='') as archivo:
            for i, tran in enumerate(transacciones, start=1):
                archivo.write(f"{i}. Tipo: {tran['tipo']}, Monto: ${tran['monto']}, Descripción: {tran['descrip']}, Categoría: {tran['categ']}\n")
        print("Exportación completada.")
    except Exception as e:
        print(f"Ocurrió un error al exportar: {type(e).__name__} - {e}")

def backUpData(transacciones):
    saveData(transacciones, "backup.csv")

def main():
    transacciones = loadData()
    while True:
        opcion = getMenuOption()
        if opcion == 1:
            addTransaction(transacciones, "I")
        elif opcion == 2:
            addTransaction(transacciones, "G")
        elif opcion == 3:
            showSummaryReport(transacciones)
            ver_submenu = input("\n¿Desea acceder al submenú? (s/n): ").lower()
            if ver_submenu == "s":
                sub_opcion = getSubMenuOption()
                if sub_opcion == 1:
                    exportData(transacciones)
                elif sub_opcion == 2:
                    backUpData(transacciones)
        elif opcion == 4:
            saveData(transacciones)
            print("Saliendo del programa... ¡Hasta la próxima!")
            break

if __name__ == "__main__":
    main()
```

### 2. Agenda de Contactos Inteligente

***Funcionalidades adicionales:***

* Almacenar contactos en un archivo CSV
* Cargar contactos automáticamente al iniciar el programa
* Implementar try-except para:
  - Validar formato de email y teléfono
  - Manejar errores de duplicados (contactos con mismo nombre/teléfono)
  - Gestionar errores de archivo (no encontrado, permisos, etc.)
  - Verificar que los campos obligatorios no estén vacíos 
* Opción de importar/exportar contactos desde/archivo
* Crear backups automáticos de la agenda

```python
import csv
from collections import Counter
import re

FILENAME = "backContact.csv"

def getMenuOption():
    while True:
        try:
            print("\n=== AGENDA DE CONTACTOS ===")
            print("1) Agregar contacto")
            print("2) Buscar contacto")
            print("3) Editar contacto existente")
            print("4) Eliminar contacto")
            print("5) Mostrar todos (ordenados alfabéticamente)")
            print("6) Contador por categoría")
            print("7) Salir")
            opcion = int(input("> "))
            if opcion in range(1, 8):
                return opcion
            else:
                print("Opción fuera de rango.")
        except Exception as e:
            print(f"Ocurrió un error: {type(e).__name__} - {e}")

def isValidEmail(email):
    ### Mi Formato para los emails: *@.*.* --> admin@gmial.com
    return bool(re.match(r"[^@]+@[^@]+\.[^@]+", email))

def isValidPhoneNumber(telefono):
    return telefono.isdigit() and len(telefono) >= 10

def isDuplicated(contactos, nombre, tel):
    for c in contactos:
        if c['nombre'].lower() == nombre.lower() or c['tel'] == tel:
            return True
    return False

def forcedInput(mensaje):
    while True:
        valor = input(mensaje).strip()
        if valor:
            return valor
        print("Este campo es obligatorio.")

def addContact(contactos, global_id):
    print("=> Agregando contacto")

    nombre = forcedInput("Nombre: ")
    tel = forcedInput("Teléfono: ")

    while not isValidPhoneNumber(tel):
        print("Teléfono inválido.")
        tel = forcedInput("Teléfono: ")

    email = forcedInput("Email: ")
    while not isValidEmail(email):
        print("Email inválido. Intenta con un formato válido (ej. usuario@dominio.com).")
        email = forcedInput("Email: ")

    categ = forcedInput("Categoría (ej: familia, trabajo, amigos): ")

    if isDuplicated(contactos, nombre, tel):
        print("Ya existe un contacto con ese nombre o teléfono.")
        return global_id  # no se incrementa

    contacto = {
        "id": global_id,
        "nombre": nombre,
        "tel": tel,
        "email": email,
        "categ": categ
    }

    contactos.append(contacto)
    print("Contacto agregado exitosamente.")
    return global_id + 1

def getSubMenuOption():
    while True:
        try:
            print("\n=== BÚSQUEDA DE CONTACTOS ===")
            print("1) Buscar por nombre")
            print("2) Buscar por categoría")
            print("3) Volver al menú")
            opcion = int(input("> "))
            if opcion in [1, 2, 3]:
                return opcion
            else:
                print("Opción fuera de rango.")
        except Exception as e:
            print(f"Ocurrió un error: {type(e).__name__} - {e}")

def printContact(c):
    print(f"ID: {c['id']} | Nombre: {c['nombre']} | Tel: {c['tel']} | Email: {c['email']} | Categoría: {c['categ']}")

def searchContact(contactos):
    while True:
        submenu = getSubMenuOption()
        if submenu == 1:
            nombre = input("Ingrese el nombre a buscar: ").lower()
            resultados = [c for c in contactos if nombre in c['nombre'].lower()]
        elif submenu == 2:
            categ = input("Ingrese la categoría a buscar: ").lower()
            resultados = [c for c in contactos if categ in c['categ'].lower()]
        else:
            break

        if resultados:
            for i, c in enumerate(resultados, 1):
                print(f"{i}. ", end="")
                printContact(c)
        else:
            print("No se encontraron resultados.")

def editContact(contactos):
    print("=> Editar un contacto")
    try:
        idx = int(input("Ingrese el ID del contacto a editar: "))
    except ValueError:
        print("ID inválido.")
        return

    for c in contactos:
        if c['id'] == idx:
            print("Contacto encontrado:")
            printContact(c)
            print("=>Escriba la nueva información o pulse un enter para saltar a otro subcampo")
            c['nombre'] = input(f"Nuevo nombre [{c['nombre']}]: ") or c['nombre']
            c['tel'] = input(f"Nuevo teléfono [{c['tel']}]: ") or c['tel']
            c['email'] = input(f"Nuevo email [{c['email']}]: ") or c['email']
            c['categ'] = input(f"Nueva categoría [{c['categ']}]: ") or c['categ']
            print("Contacto actualizado:")
            printContact(c)
            return
    print("No se encontró contacto con ese ID.")

def deleteContact(contactos):
    print("=> Eliminar un contacto")
    try:
        idx = int(input("Ingrese el ID del contacto a eliminar: "))
    except ValueError:
        print("ID inválido.")
        return

    for i, c in enumerate(contactos):
        if c['id'] == idx:
            print("Contacto encontrado:")
            printContact(c)
            confirm = input("¿Estás seguro de eliminar este contacto? (s/n): ").lower()
            if confirm == "s":
                contactos.pop(i)
                print("Contacto eliminado.")
            return
    print("No se encontró contacto con ese ID.")

def sortedContact(contactos):
    print("=> Contactos ordenados alfabéticamente")
    if not contactos:
        print("No hay contactos para mostrar.")
        return
    for c in sorted(contactos, key=lambda d: d['nombre'].lower()):
        printContact(c)

def showCategoriesData(contactos):
    if not contactos:
        print("No hay contactos.")
        return
    contador = Counter(c['categ'] for c in contactos)
    for cat, count in contador.items():
        print(f"Categoría: {cat} | Cantidad: {count}")

def saveData(contactos, filename=FILENAME):
    print(f"Guardando en {filename}...")
    try:
        with open(filename, "w", newline='', encoding='utf-8') as archivo:
            campos = ["id", "nombre", "tel", "email", "categ"]
            writer = csv.DictWriter(archivo, fieldnames=campos)
            writer.writeheader()
            writer.writerows(contactos)
        print("Datos guardados.")
    except Exception as e:
        print(f"Error al guardar: {type(e).__name__} - {e}")

def loadData(filename=FILENAME):
    print(f"Cargando datos desde {filename}...")
    contactos = []
    idx = 0
    try:
        with open(filename, "r", encoding='utf-8') as archivo:
            reader = csv.DictReader(archivo)
            for fila in reader:
                fila["id"] = int(fila["id"])
                idx = max(idx, fila["id"])
                contactos.append(fila)
        print("Datos cargados correctamente.")
    except FileNotFoundError:
        print("Archivo no encontrado. Se usará una lista vacía.")
    except Exception as e:
        print(f"Error al cargar: {type(e).__name__} - {e}")
    return contactos, idx + 1

# === PRINCIPAL ===
contactos, global_id = loadData()

while True:
    opcion = getMenuOption()

    if opcion == 1:
        global_id = addContact(contactos, global_id)
        saveData(contactos)

    elif opcion == 2:
        searchContact(contactos)

    elif opcion == 3:
        editContact(contactos)
        saveData(contactos)

    elif opcion == 4:
        deleteContact(contactos)
        saveData(contactos)

    elif opcion == 5:
        sortedContact(contactos)

    elif opcion == 6:
        showCategoriesData(contactos)

    elif opcion == 7:
        print("Saliendo...")
        saveData(contactos)
        break
```

### 3. Sistema de inventario para tienda

***Funcionalidades adicionales:*** 

* Guardar el inventario completo en un archivo CSV.
* Cargar inventario automáticamente al iniciar el programa.
* Implementar try-except para:
  - Validar que precio y cantidad sean valores numéricos válidos.
  - Manejar errores de stock insuficiente al realizar ventas.
  - Gestionar errores de archivo (corrupción, permisos, etc.).
  - Verificar que no existan productos duplicados.
* Generar reportes en archivos de texto.
* Mantener un historial de movimientos (entradas/salidas) en un archivo de log.
* Opción para realizar backup del inventario.

```python
import csv

FILENAME = "inventario.csv"

# === UTILITARIOS ===

def forcedInput(mensaje):
    while True:
        valor = input(mensaje).strip()
        if valor:
            return valor
        print("Este campo es obligatorio.")

def getInputFloat(mensaje="Ingresa un número decimal: "):
    while True:
        valor = forcedInput(mensaje)
        try:
            return float(valor)
        except ValueError:
            print("Valor no válido. Intenta de nuevo.")

def getInputInt(mensaje="Ingresa un número entero: "):
    while True:
        valor = forcedInput(mensaje)
        try:
            return int(valor)
        except ValueError:
            print("Valor no válido. Intenta de nuevo.")

def log(mensaje):
    try:
        with open("log.txt", "a", encoding='utf-8') as archivo:
            archivo.write(mensaje + "\n")
        print("[LOG] " + mensaje)
    except Exception as e:
        print(f"[ERROR LOG] {type(e).__name__} - {e}")

def printProduct(p):
    return f"ID: {p['id']} | Nombre: {p['nombre']} | Precio: ${p['precio']:.2f} | Cantidad: {p['cantidad']}"

def isDuplicated(productos, nombre):
    for p in productos:
        if p['nombre'].lower() == nombre.lower():
            return True
    return False

# === FUNCIONES PRINCIPALES ===

def getMenuOption():
    while True:
        try:
            print("\n=== MENÚ DE INVENTARIO ===")
            print("1) Registrar productos")
            print("2) Registrar venta")
            print("3) Generar reporte")
            print("4) Realizar BackUp")
            print("5) Salir")
            opcion = getInputInt("> ")
            if 1 <= opcion <= 5:
                return opcion
            else:
                print("Opción fuera de rango.")
        except ValueError:
            print("Entrada no válida. Debes ingresar un número.")

def addProduct(productos, global_id):
    print("\n=== Registrar nuevo producto ===")
    nombre = forcedInput("Nombre del producto: ")

    if isDuplicated(productos, nombre):
        print("Ya existe un producto con ese nombre.")
        return global_id

    precio = -1
    while precio <= 0:
        precio = getInputFloat("Precio del producto: ")
        if precio <= 0:
            print("El precio debe ser mayor que cero.")

    cantidad = -1
    while cantidad <= 0:
        cantidad = getInputInt("Cantidad en stock: ")
        if cantidad <= 0:
            print("La cantidad debe ser mayor que cero.")

    producto = {
        "id": global_id,
        "nombre": nombre,
        "precio": precio,
        "cantidad": cantidad
    }

    productos.append(producto)
    print("Producto agregado exitosamente.")
    log("[ALTA] " + printProduct(producto))
    return global_id + 1

def sellProduct(productos):
    print("\n=== Registrar venta de producto ===")
    nombre = forcedInput("Nombre del producto a vender: ").lower()
    idx = next((i for i, p in enumerate(productos) if p['nombre'].lower() == nombre), -1)

    if idx == -1:
        print("Producto no encontrado.")
        return

    print("Producto encontrado:")
    print(printProduct(productos[idx]))

    cantidad = -1
    while cantidad <= 0:
        cantidad = getInputInt("Cantidad a vender: ")
        if cantidad <= 0:
            print("La cantidad debe ser mayor que cero.")

    if productos[idx]["cantidad"] < cantidad:
        print("Stock insuficiente.")
        return

    productos[idx]["cantidad"] -= cantidad
    log(f"[VENTA] Se vendieron {cantidad} unidades de '{productos[idx]['nombre']}'.")

    if productos[idx]["cantidad"] == 0:
        log(f"[AGOTADO] Producto '{productos[idx]['nombre']}' agotado y eliminado del inventario.")
        productos.pop(idx)

def generateReport(filename, productos):
    print("\n=== Generando Reporte ===")
    try:
        with open(filename, "w", encoding='utf-8') as archivo:
            total = 0
            for i, p in enumerate(productos):
                linea = f"{i+1}. {printProduct(p)}\n"
                archivo.write(linea)
                total += p["precio"] * p["cantidad"]
            archivo.write(f"\nTOTAL EN STOCK: ${total:.2f}\n")
        print(f" Reporte generado en '{filename}'")
    except Exception as e:
        print(f"[ERROR REPORTE] {type(e).__name__} - {e}")

def saveData(productos, filename=FILENAME):
    print(f"\n Guardando en {filename}...")
    try:
        with open(filename, "w", newline='', encoding='utf-8') as archivo:
            campos = ["id", "nombre", "precio", "cantidad"]
            writer = csv.DictWriter(archivo, fieldnames=campos)
            writer.writeheader()
            writer.writerows(productos)
        print(" Datos guardados.")
    except Exception as e:
        print(f"[ERROR GUARDADO] {type(e).__name__} - {e}")

def loadData(filename=FILENAME):
    print(f"\n Cargando datos desde {filename}...")
    productos = []
    global_id = 1
    try:
        with open(filename, "r", encoding='utf-8') as archivo:
            reader = csv.DictReader(archivo)
            for fila in reader:
                fila["id"] = int(fila["id"])
                fila["precio"] = float(fila["precio"])
                fila["cantidad"] = int(fila["cantidad"])
                productos.append(fila)
                global_id = max(global_id, fila["id"] + 1)
        print(" Datos cargados correctamente.")
    except FileNotFoundError:
        print("⚠️ Archivo no encontrado. Iniciando inventario vacío.")
    except Exception as e:
        print(f"[ERROR CARGA] {type(e).__name__} - {e}")
    return productos, global_id

def backUpData(productos):
    print("\n Realizando copia de seguridad...")
    saveData(productos, "backupProductos.csv")

# === PROGRAMA PRINCIPAL ===

productos, global_id = loadData()

while True:
    opcion = getMenuOption()

    if opcion == 1:
        global_id = addProduct(productos, global_id)
    elif opcion == 2:
        sellProduct(productos)
    elif opcion == 3:
        generateReport("reporte.txt", productos)
    elif opcion == 4:
        backUpData(productos)
    elif opcion == 5:
        print("\n Saliendo del programa...")
        saveData(productos)
        break
```
