### 1.Simulador de Gestión de Finanzas Personales

**Descripción**: 

*Crear un programa que permita a un usuario registrar sus ingresos y gastos, categorizarlos y generar un resumen de su situación financiera.*

El programa debe: 
* Agregar ingreso (monto, descripción, categoría)
* Agregar gasto (monto, descripción, categoría)
* Ver resumen (balance actual, gastos por categoría)
* Calcular balance total (suma ingresos - suma gastos)
* Mostrar distribución de gastos por categoría

```python
### Creamos nuestra lista la cual tendrá todos nuestro movimientos.
transacciones = []
runTask = True
while runTask:
    print("===MENU===")
    opcion = int(input("1) Agregar Ingreso\n2) Agregar Gasto\n3) Ver Resumen\n4) Salir\n"))

    if opcion == 1:
        movimiento = {
            "tipo": "I",
            "monto": 0.0,
            "descrip": "",
            "categ": ""
        }

        montoValido = False
        while not montoValido:
            monto = float(input("Ingrese el monto a ingresar: "))
            if monto < 0:
                print("No se pueden montos negativos")
            else:
                montoValido = True
        movimiento["monto"] = monto

        descrip = input("Ingrese la descripcion del ingreso: ")
        movimiento["descrip"] = descrip

        categ = input("Ingrese la categoría de este movimiento: ")
        movimiento["categ"] = categ

        transacciones.append(movimiento)

    elif opcion == 2:
        movimiento = {  # nuevo diccionario para cada gasto
            "tipo": "G",
            "monto": 0.0,
            "descrip": "",
            "categ": ""
        }

        montoValido = False
        while not montoValido:
            monto = float(input("Ingrese el monto a gastar: "))
            if monto > 0:
                print("No se pueden montos positivos (usa valores negativos)")
            else:
                montoValido = True
        movimiento["monto"] = monto

        descrip = input("Ingrese la descripcion del gasto: ")
        movimiento["descrip"] = descrip

        categ = input("Ingrese la categoría de este movimiento: ")
        movimiento["categ"] = categ

        transacciones.append(movimiento)

    elif opcion == 3:
        total = 0
        gasto_categoria = {}
        for transaccion in transacciones:
            total += transaccion["monto"]
            if transaccion["categ"] in gasto_categoria:
                gasto_categoria[transaccion["categ"]] += transaccion["monto"]
            else:
                gasto_categoria[transaccion["categ"]] = transaccion["monto"]

        print(f"Balance total: ${total}")
        for categoria in gasto_categoria.keys():
            print(f"Categoria: [{categoria}] -> ${gasto_categoria[categoria]}")

    elif opcion == 4:
        runTask = False
        print("Saliendo del programa...")
    else:
        print("Opción inválida, intenta de nuevo")
```

### 2. Agenda de Contactos Inteligente 

**Descripción**: 

*Crear una agenda de contactos que permita gestionar información de personas.* 

El programa debe: 
* Añadir contactos (nombre, teléfono, email, categoría)
* Buscar contactos por nombre o categoría
* Editar información de contactos existentes
* Eliminar contactos
* Mostrar todos los contactos ordenados alfabéticamente
* Contador de contactos por categoría (familia, trabajo, amigos)

```python
### Definimos nuestros lista con los registros.
contactos = []
global_id = 1

while True:
    print("\n=== AGENDA DE CONTACTOS ===")
    print("1) Agregar contacto")
    print("2) Buscar contacto")
    print("3) Editar contacto existente")
    print("4) Eliminar contacto")
    print("5) Mostrar todos (ordenados alfabéticamente)")
    print("6) Contador por categoría")
    print("7) Salir")
    opcion = int(input("Elija una opción (1-7): "))
    if opcion == 1:
        print("=> Agregando contacto")
        nombre = input("Nombre: ")
        tel = input("Teléfono: ")
        email = input("Email: ")
        categ = input("Categoría (ej: familia, trabajo, amigos): ")
        contacto = {
            "id": global_id,
            "nombre": nombre,
            "tel": tel,"email": email,
            "categ": categ
        }
        contactos.append(contacto)
        print("Contacto agregado")
        global_id += 1

    elif opcion == 2:
        print("=> Buscar contacto")
        while True:
            submenu = int(input("Buscar por: 1) Nombre  2) Categoría  3) Volver\nElija: "))
            if submenu == 1:
                nombre = input("Ingrese el nombre a buscar: ")
                ### Usamos una expresion para no tener tantas lineas
                resultados = [contacto for contacto in contactos if nombre == contacto['nombre']]
                if resultados:
                    for i, c in enumerate(resultados, 1):
                        print(f"{i}. ", end="")
                        print(f"ID: {c['id']} | Nombre: {c['nombre']} | Tel: {c['tel']} | Email: {c['email']} | Categoría: {c['categ']}")
                else:
                    print(f"No se encontraron contactos con nombre que contenga: '{nombre}'")
        
            elif submenu == 2:
                categ = input("Ingrese la categoría a buscar: ")
                resultados = [contacto for contacto in contactos if categ == contacto['categ']]
                if resultados:
                    for i, c in enumerate(resultados, 1):
                        print(f"{i}. ", end="")
                        print(f"ID: {c['id']} | Nombre: {c['nombre']} | Tel: {c['tel']} | Email: {c['email']} | Categoría: {c['categ']}")
                else:
                    print(f"No se encontraron contactos en la categoría: '{categ}'")
            elif submenu == 3:
                break
            else:
                print("Opción inválida del submenú.")
            
    elif opcion == 3:
        print("=> Editar Un contacto.")
        idx = int(input("Ingrese el ID del registro a modificar: "))
        for c in contactos:
            if c['id'] == idx:
                print("Registro encontrado:")
                print(f"ID: {c['id']} | Nombre: {c['nombre']} | Tel: {c['tel']} | Email: {c['email']} | Categoría: {c['categ']}")
                nuevo_nombre = input(f"Nuevo nombre [{c['nombre']}]: ")
                nuevo_tel = input(f"Nuevo teléfono [{c['tel']}]: ")
                nuevo_email = input(f"Nuevo email [{c['email']}]: ")
                nueva_categ = input(f"Nueva categoría [{c['categ']}]: ")
                c['nombre'] = nuevo_nombre
                c['tel'] = nuevo_tel
                c['email'] = nuevo_email
                c['categ'] = nueva_categ
                print("Registro actualizado:")
                print(f"ID: {c['id']} | Nombre: {c['nombre']} | Tel: {c['tel']} | Email: {c['email']} | Categoría: {c['categ']}")
                break

    elif opcion == 4:
        print("=> Eliminar un contacto")
        idx = int(input("Ingrese el ID del registro a eliminar: "))
        for i, c in enumerate(contactos):
            if c['id'] == idx:
                print("Registro encontrado:")
                print(f"ID: {c['id']} | Nombre: {c['nombre']} | Tel: {c['tel']} | Email: {c['email']} | Categoría: {c['categ']}")
                contactos.pop(i)
                print("Contacto eliminado.")
                break
    elif opcion == 5:
        print("=> Ordenando Alfabeticamente")
        if not contactos:
            print("No hay contactos para mostrar.")
            continue
        ordenados = sorted(contactos, key=lambda d: d['nombre'])
        for c in ordenados:
            print(f"ID: {c['id']} | Nombre: {c['nombre']} | Tel: {c['tel']} | Email: {c['email']} | Categoría: {c['categ']}")               
    elif opcion == 6:
        if not contactos:
            print("No hay contactos.")
            continue
        ### Sacar todas las diferentes categorias
        categorias = {}
        for contacto in contactos: 
            if contacto['categ'] not in categorias.keys():
                categorias[contacto['categ']] = 1
            else:
                categorias[contacto['categ']]+=1
        
        for cat, count in categorias.items():
            print(f" Categoría: {cat} Apariciones: {count}") 
    elif opcion == 7:
            print("Saliendo...")
            break
    else:
            print("Opción inválida, intenta de nuevo.")
```

### 3. Simulador de Caja Registradora Simple 

**Descripción**: 

*Crea un simulador de caja registradora para sumar productos*. 

El programa debe: 
- Permitir ingresar precios de productos uno por uno
- Ir mostrando el subtotal después de cada producto
- Preguntar después de cada producto si se desea agregar otro
- Calcular el total final cuando el usuario decida terminar
- Mostrar el total con formato de dinero

```python
### Hasta que el usuario quiera
total = 0
while True:
    print(f"Total Actual: ${total:,.2f}")
    precio=float(input("Ingrese el precio del producto"))
    total += precio
    opcion=int(input("Desea continuara?\n1) Si\n2) No\n>:"))
    if opcion != 1 :
        break
print(f"Total Final: ${total:,.2f}")
```

### 4. Juego: "Pares o Nones" con Dados

Descripción: 
Un juego clásico donde dos jugadores apuestan si la suma de dos dados será par o 
impar. Gana quien acierte el resultado. 
Cómo funciona: 
1. Cada jugador elige "par" o "impar" 
2. Se lanzan dos dados virtuales 
3. Se suma el resultado de los dados 
4. Gana el jugador que haya elegido correctamente si la suma es par o impar

```python
import random

hayGanador = False
while not hayGanador:
    print("\nCARRRERA DE DADOS!")
    j1 = int(input("\n[JUGADOR1]\n1)PAR\n2)IMPAR\n>"))
    j2 = int(input("\n[JUGADOR2]\n1)PAR\n2)IMPAR\n>"))
    dado1 = random.randint(1,6)
    dado2 = random.randint(1,6)
    suma = dado1+dado2
    print(f"Dado1: {dado1}, Dado2: {dado2} --> Suma {suma}")
    esPar = True if suma%2 == 0 else False
    if esPar:
        if j1==1 and j2==1 :
            print("\nAMBOS JUGADORES HAN ACERTADO!")
            hayGanador = True
            break
        if j1==1:
            print("\nEL JUGADOR 1 HA ACERTADO!")
            hayGanador = True
            break
        if j2==1:
            print("\nEL JUGADOR 2 HA ACERTADO!")
            hayGanador = True
            break
        print("\nNINGUN JUGADOR ACERTÓ!")
    else:
        if j1==2 and j2==2:
            print("\nAMBOS JUGADORES HAN ACERTADO!")
            hayGanador = True
            break
        if j1==2:
            print("\nEL JUGADOR 1 HA ACERTADO!")
            hayGanador = True
            break
        if j2==2:
            print("\nEL JUGADOR 2 HA ACERTADO!")
            hayGanador = True
            break
        print("\nNINGUN JUGADOR ACERTÓ!")
```
