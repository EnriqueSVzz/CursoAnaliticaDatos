#  Diccionarios en Python

##  Objetivo de la clase
Aprender a crear, manipular y usar diccionarios en Python para almacenar datos de forma organizada.




## Introduccion a diccionarios

### 1.  ¿Qué es un diccionario?

Un diccionario es una estructura de datos que almacena información en pares **clave-valor**.

**Características:**

* **Claves:** Únicas y sirven como "identificadores"
* **Valores:** Pueden ser cualquier tipo de dato
* **No ordenado:** No guardan un orden específico (en versiones antiguas de Python)
* **Rápido acceso:** Encontrar valores por clave es muy rápido
```python
# Así se ve un diccionario
persona = {
    "nombre": "María",
    "edad": 28,
    "ciudad": "Madrid",
    "hobbies": ["leer", "nadar", "cocinar"]
}



````python
# Creando nuestro primer diccionario
print("=== CREANDO DICCIONARIOS ===")

# Diccionario de una persona
persona = {
    "nombre": "Carlos",
    "edad": 25,
    "ciudad": "Barcelona",
    "casado": False
}

print("Diccionario completo:", persona)
````

### 2. Acceder a los valores

Podemos acceder a los valores usando sus claves:

````python
# Accediendo a valores del diccionario
print("=== ACCEDIENDO A VALORES ===")

print("Nombre:", persona["nombre"])
print("Edad:", persona["edad"])
print("Ciudad:", persona["ciudad"])

# También podemos usar .get() que es más seguro
print("¿Casado?:", persona.get("casado"))
print("Profesión:", persona.get("profesion", "No especificada"))  # Valor por defecto

# Intentar acceder a una clave que no existe (da error)
try:
    print("Profesión:", persona["profesion"])
except KeyError as e:
    print(f"Error: La clave {e} no existe en el diccionario")
````

### 3. Modificar y agregar valores

Los diccionarios son mutables, podemos cambiar sus valores:

````python
# Modificando valores
print("=== MODIFICANDO VALORES ===")

print("Diccionario original:", persona)

# Cambiar valores existentes
persona["edad"] = 26
persona["ciudad"] = "Valencia"

# Agregar nuevos valores
persona["profesion"] = "Ingeniero"
persona["email"] = "carlos@email.com"

print("Diccionario modificado:", persona)
````

### 4. Eliminar elementos

Podemos eliminar elementos de varias formas:

````python
# Eliminando elementos
print("=== ELIMINANDO ELEMENTOS ===")

print("Diccionario antes de eliminar:", persona)

# Eliminar con del
del persona["casado"]
print("Después de del persona['casado']:", persona)

# Eliminar con pop() (devuelve el valor)
email = persona.pop("email")
print(f"Email eliminado: {email}")
print("Después de pop('email'):", persona)

# Eliminar con popitem() (elimina el último item)
ultimo = persona.popitem()
print(f"Último item eliminado: {ultimo}")
print("Después de popitem():", persona)

# Eliminar todos los elementos
persona.clear()
print("Después de clear():", persona)
````

### 5. Métodos útiles de diccionarios

Los diccionarios tienen muchos métodos útiles:

````python
# Métodos útiles de diccionarios
print("=== MÉTODOS ÚTILES ===")

# Recreamos nuestro diccionario
persona = {
    "nombre": "Ana",
    "edad": 30,
    "ciudad": "Sevilla",
    "hobbies": ["pintar", "bailar"]
}

print("Diccionario:", persona)

# Obtener todas las claves
print("Claves:", persona.keys())

# Obtener todos los valores
print("Valores:", persona.values())

# Obtener todos los items (pares clave-valor)
print("Items:", persona.items())

# Copiar un diccionario
copia_persona = persona.copy()
print("Copia:", copia_persona)

# Verificar si una clave existe
print("¿Existe 'nombre'?", "nombre" in persona)
print("¿Existe 'telefono'?", "telefono" in persona)
````

### 6. Recorrer un diccionario

Podemos recorrer diccionarios de varias formas:

````python
# Recorriendo diccionarios
print("=== RECORRIENDO DICCIONARIOS ===")

print("Recorriendo solo claves:")
for clave in persona:
    print(f"- {clave}")

print("\nRecorriendo claves y valores:")
for clave, valor in persona.items():
    print(f"- {clave}: {valor}")

print("\nRecorriendo solo valores:")
for valor in persona.values():
    print(f"- {valor}")
````

### 7. Diccionarios anidados

Los diccionarios pueden contener otros diccionarios:

````python
# Diccionarios anidados
print("=== DICCIONARIOS ANIDADOS ===")

# Diccionario más complejo
usuario = {
    "datos_personales": {
        "nombre": "Laura",
        "edad": 28,
        "direccion": {
            "calle": "Av. Principal",
            "numero": 123,
            "ciudad": "Madrid"
        }
    },
    "preferencias": {
        "color_favorito": "azul",
        "comida_favorita": "pizza",
        "hobbies": ["leer", "viajar", "fotografía"]
    },
    "estado": "activo"
}

print("Diccionario anidado completo:")
print(usuario)

# Acceder a valores anidados
print("\nAccediendo a valores anidados:")
print("Nombre:", usuario["datos_personales"]["nombre"])
print("Calle:", usuario["datos_personales"]["direccion"]["calle"])
print("Hobbies:", usuario["preferencias"]["hobbies"])
````

## 8. Ejercicios de práctica

Practiquemos con algunos ejercicios:

````python
# Ejercicio 1: Crear un diccionario de libro
print("=== EJERCICIO 1: DICCIONARIO DE LIBRO ===")

libro = {
    "titulo": "Cien años de soledad",
    "autor": "Gabriel García Márquez",
    "año": 1967,
    "genero": "Realismo mágico",
    "paginas": 432
}

# Tu código: muestra toda la información del libro
print("Información del libro:")
info = input("¿Que información neceitas saber?")
print(f"La información solicitada es {info}:")
try:
    print(f"\n {libro[info]}")
except KeyError as e:
    print(f"Error: No existe registro de esa información")
````

````python
# Ejercicio 2: Agenda telefónica
print("\n=== EJERCICIO 2: AGENDA TELEFÓNICA ===")

agenda = {
    "Juan": "123-456-789",
    "Maria": "987-654-321",
    "Carlos": "555-123-456",
    "Ana":"777-888-999",
    "Luis":"111-222-333"
}
verificador = True
while verificador:
  info = input("¿Que accion te gustaria realizar? (Mostrar todos/Buscar/Salir)")

  if info=="Mostrar todos":
    # Tu código: mostrar todos los contactos
    print("Contactos en agenda:")
    for nombre, telefono in agenda.items():
        print(f"{nombre}: {telefono}")
  elif info=="Buscar":
    # Tu código: buscar un contacto específico
    buscar = input("¿Que numero qusieras ver?")
    if buscar in agenda:
        print(f"\nTeléfono de {buscar}: {agenda[buscar]}")
    else:
        print(f"\n{buscar} no está en la agenda")
        continue
  elif info=="Salir":
    print("¡Hasta luego!")
    verificador = False
  else:
    print(f"\n Opción no valida")
    continue
````

````python
# Ejercicio 4: Estudio de Ecosistemas - Biodiversidad de un Bosque
print("=== EJERCICIO 4: ESTUDIO DE ECOSISTEMAS ===")
print("Análisis de biodiversidad en un bosque tropical\n")

# Diccionario que representa las especies de un bosque
bosque_tropical = {
    "plantas": {
        "árboles": {
            "Caoba": {
                "cantidad": 45,
                "altura_promedio": 25.5,
                "edad_promedio": 85,
                "estado_conservacion": "vulnerable"
            },
            "Cedro": {
                "cantidad": 32,
                "altura_promedio": 22.8,
                "edad_promedio": 70,
                "estado_conservacion": "en peligro"
            },
            "Pino": {
                "cantidad": 78,
                "altura_promedio": 18.3,
                "edad_promedio": 45,
                "estado_conservacion": "preocupación menor"
            }
        },
        "arbustos": {
            "total_especies": 15,
            "cobertura_porcentaje": 25.7
        },
        "hierbas": {
            "total_especies": 42,
            "cobertura_porcentaje": 18.3
        }
    },
    "animales": {
        "mamiferos": {
            "Jaguar": {
                "poblacion_estimada": 8,
                "estado_conservacion": "casi amenazado",
                "dieta": "carnívoro"
            },
            "Mono araña": {
                "poblacion_estimada": 35,
                "estado_conservacion": "en peligro",
                "dieta": "frugívoro"
            },
            "Oso hormiguero": {
                "poblacion_estimada": 12,
                "estado_conservacion": "vulnerable",
                "dieta": "insectívoro"
            }
        },
        "aves": {
            "total_especies": 67,
            "aves_migratorias": 23,
            "aves_endemicas": 15
        },
        "reptiles": {
            "total_especies": 28,
            "serpientes_venenosas": 5
        }
    },
    "caracteristicas_ecosistema": {
        "area_hectareas": 500,
        "precipitacion_anual_mm": 2500,
        "temperatura_promedio_c": 26.5,
        "estacion_lluviosa": "mayo a octubre"
    }
}

# Análisis 1: Información general del bosque
print("1. 📊 INFORMACIÓN GENERAL DEL BOSQUE")
print(f"Área total: {bosque_tropical['caracteristicas_ecosistema']['area_hectareas']} hectáreas")
print(f"Precipitación anual: {bosque_tropical['caracteristicas_ecosistema']['precipitacion_anual_mm']} mm")
print(f"Temperatura promedio: {bosque_tropical['caracteristicas_ecosistema']['temperatura_promedio_c']}°C")
print(f"Estación lluviosa: {bosque_tropical['caracteristicas_ecosistema']['estacion_lluviosa']}\n")

# Análisis 2: Biodiversidad de plantas
print("2. 🌿 BIODIVERSIDAD DE PLANTAS")
total_arboles = sum(arbol['cantidad'] for arbol in bosque_tropical['plantas']['árboles'].values())
print(f"Total de árboles registrados: {total_arboles}")
print(f"Especies de arbustos: {bosque_tropical['plantas']['arbustos']['total_especies']}")
print(f"Especies de hierbas: {bosque_tropical['plantas']['hierbas']['total_especies']}\n")

# Análisis 3: Especies en peligro
print("3. ⚠️ ESPECIES EN PELIGRO DE EXTINCIÓN")
print("Especies de árboles en estado crítico:")
for nombre_arbol, info_arbol in bosque_tropical['plantas']['árboles'].items():
    if info_arbol['estado_conservacion'] in ['en peligro', 'vulnerable']:
        print(f"  - {nombre_arbol}: {info_arbol['estado_conservacion']}")

print("\nEspecies de animales en estado crítico:")
for nombre_animal, info_animal in bosque_tropical['animales']['mamiferos'].items():
    if info_animal['estado_conservacion'] in ['en peligro', 'vulnerable', 'casi amenazado']:
        print(f"  - {nombre_animal}: {info_animal['estado_conservacion']}")

# Análisis 4: Cálculo de densidad poblacional
print("\n4. 📈 DENSIDAD POBLACIONAL")
area = bosque_tropical['caracteristicas_ecosistema']['area_hectareas']
print(f"Densidad de árboles: {total_arboles/area:.2f} árboles/hectárea")

total_mamiferos = sum(mamifero['poblacion_estimada'] for mamifero in bosque_tropical['animales']['mamiferos'].values())
print(f"Densidad de mamíferos: {total_mamiferos/area:.2f} mamíferos/hectárea")

# Análisis 5: Diversidad de aves
print("\n5. 🐦 DIVERSIDAD DE AVES")
aves = bosque_tropical['animales']['aves']
print(f"Total de especies de aves: {aves['total_especies']}")
print(f"Aves migratorias: {aves['aves_migratorias']}")
print(f"Aves endémicas: {aves['aves_endemicas']}")
print(f"Porcentaje de aves endémicas: {(aves['aves_endemicas']/aves['total_especies'])*100:.1f}%")

# Tu turno: Agregar una nueva especie al bosque
print("\n6. 🆕 AGREGAR NUEVA ESPECIE (Tu ejercicio)")

# Ejemplo: Agregar un nuevo árbol - Guayacán
bosque_tropical['plantas']['árboles']['Guayacán'] = {
    "cantidad": 27,
    "altura_promedio": 20.1,
    "edad_promedio": 60,
    "estado_conservacion": "vulnerable"
}

# Ejemplo: Agregar un nuevo mamífero - Puma
bosque_tropical['animales']['mamiferos']['Puma'] = {
    "poblacion_estimada": 6,
    "estado_conservacion": "casi amenazado",
    "dieta": "carnívoro"
}

print("Nuevas especies agregadas: Guayacán (árbol) y Puma (mamífero)")

# Mostrar el diccionario actualizado (solo parte de él para no saturar)
print(f"\nTotal de especies de árboles ahora: {len(bosque_tropical['plantas']['árboles'])}")
print(f"Total de especies de mamíferos ahora: {len(bosque_tropical['animales']['mamiferos'])}")

# Búsqueda de información específica
print("\n7. 🔍 BÚSQUEDA DE INFORMACIÓN ESPECÍFICA")
especie_buscar = "Jaguar"
if especie_buscar in bosque_tropical['animales']['mamiferos']:
    info_jaguar = bosque_tropical['animales']['mamiferos'][especie_buscar]
    print(f"Información del {especie_buscar}:")
    print(f"  - Población estimada: {info_jaguar['poblacion_estimada']}")
    print(f"  - Estado de conservación: {info_jaguar['estado_conservacion']}")
    print(f"  - Dieta: {info_jaguar['dieta']}")
else:
    print(f"La especie {especie_buscar} no se encuentra en el registro")

# Análisis de cobertura vegetal
print("\n8. 🌳 COBERTURA VEGETAL TOTAL")
cobertura_total = (
    bosque_tropical['plantas']['arbustos']['cobertura_porcentaje'] +
    bosque_tropical['plantas']['hierbas']['cobertura_porcentaje']
)
print(f"Cobertura vegetal total: {cobertura_total}%")
print(f"Espacio no cubierto: {100 - cobertura_total}%")

print("\n¡Análisis del ecosistema completado! 🌎")
````

````python
# Ejercicio 5: Base de Datos de Elementos Radiactivos
print("=== EJERCICIO 5: FÍSICA NUCLEAR ===")
print("Base de datos de elementos radiactivos y sus propiedades\n")

# Diccionario de elementos radiactivos importantes
elementos_radiactivos = {
    "Uranio-235": {
        "simbolo": "U-235",
        "numero_atomico": 92,
        "masa_atomica": 235.0439299,
        "vida_media": 703800000,  # años
        "tipo_decaimiento": "alfa",
        "energia_decaimiento_MeV": 4.679,
        "usos": ["reactores nucleares", "armas nucleares"],
        "descubrimiento": 1935,
        "descubridor": "Arthur Jeffrey Dempster"
    },
    "Plutonio-239": {
        "simbolo": "Pu-239",
        "numero_atomico": 94,
        "masa_atomica": 239.0521634,
        "vida_media": 24100,  # años
        "tipo_decaimiento": "alfa",
        "energia_decaimiento_MeV": 5.245,
        "usos": ["reactores nucleares", "armas nucleares", "generadores termoeléctricos"],
        "descubrimiento": 1941,
        "descubridor": "Glenn T. Seaborg"
    },
    "Carbono-14": {
        "simbolo": "C-14",
        "numero_atomico": 6,
        "masa_atomica": 14.003241,
        "vida_media": 5730,  # años
        "tipo_decaimiento": "beta",
        "energia_decaimiento_MeV": 0.156,
        "usos": ["datación arqueológica", "trazadores biológicos"],
        "descubrimiento": 1940,
        "descubridor": "Martin Kamen"
    },
    "Radio-226": {
        "simbolo": "Ra-226",
        "numero_atomico": 88,
        "masa_atomica": 226.0254098,
        "vida_media": 1600,  # años
        "tipo_decaimiento": "alfa",
        "energia_decaimiento_MeV": 4.870,
        "usos": ["medicina nuclear", "investigación"],
        "descubrimiento": 1898,
        "descubridor": "Marie Curie"
    },
    "Tecnecio-99m": {
        "simbolo": "Tc-99m",
        "numero_atomico": 43,
        "masa_atomica": 98.9062547,
        "vida_media": 6.0058,  # horas
        "tipo_decaimiento": "gamma",
        "energia_decaimiento_MeV": 0.141,
        "usos": ["medicina nuclear", "diagnóstico por imagen"],
        "descubrimiento": 1937,
        "descubridor": "Carlo Perrier"
    }
}

# Análisis 1: Información general de los elementos
print("1. 📊 INFORMACIÓN GENERAL DE ELEMENTOS RADIACTIVOS")
print("Elementos en la base de datos:")
for elemento in elementos_radiactivos.keys():
    print(f"  - {elemento} ({elementos_radiactivos[elemento]['simbolo']})")

# Análisis 2: Elementos por tipo de decaimiento
print("\n2. ⚡ TIPOS DE DECAIMIENTO RADIACTIVO")
decaimientos = {}
for elemento, info in elementos_radiactivos.items():
    tipo = info['tipo_decaimiento']
    if tipo in decaimientos:
        decaimientos[tipo].append(elemento)
    else:
        decaimientos[tipo] = [elemento]

for tipo, elementos in decaimientos.items():
    print(f"Decaimiento {tipo}: {len(elementos)} elementos")
    for elem in elementos:
        print(f"  - {elem}")

# Análisis 3: Cálculo de actividad radiactiva
print("\n3. 🧮 CÁLCULO DE ACTIVIDAD RADIACTIVA")
print("Constante de decaimiento (λ) y actividad para 1 gramo:")

for elemento, info in elementos_radiactivos.items():
    # λ = ln(2) / vida_media
    vida_media = info['vida_media']
    # Convertir a segundos (asumiendo que la vida media está en años)
    vida_media_segundos = vida_media * 365.25 * 24 * 3600
    constante_decaimiento = 0.693 / vida_media_segundos

    # Número de átomos en 1 gramo
    numero_atomos = 1 / info['masa_atomica'] * 6.022e23  # Número de Avogadro
    actividad = constante_decaimiento * numero_atomos  # Bq (decaimientos/segundo)

    print(f"{elemento}: λ = {constante_decaimiento:.2e} s⁻¹, Actividad = {actividad:.2e} Bq")

# Análisis 4: Elementos descubiertos por década
print("\n4. 📅 DESCUBRIMIENTOS POR DÉCADA")
decadas = {}
for elemento, info in elementos_radiactivos.items():
    año = info['descubrimiento']
    decada = (año // 10) * 10
    if decada in decadas:
        decadas[decada].append(elemento)
    else:
        decadas[decada] = [elemento]

for decada in sorted(decadas.keys()):
    print(f"Década {decada}s: {len(decadas[decada])} elementos")
    for elem in decadas[decada]:
        print(f"  - {elem} ({elementos_radiactivos[elem]['descubridor']})")

# Análisis 5: Energía liberada por decaimiento
print("\n5. ⚡ ENERGÍA POR DECAIMIENTO")
print("Energía liberada en cada decaimiento:")
for elemento, info in elementos_radiactivos.items():
    energia_mev = info['energia_decaimiento_MeV']
    energia_joules = energia_mev * 1.602e-13  # Conversión a joules
    print(f"{elemento}: {energia_mev} MeV = {energia_joules:.2e} J")

# Tu turno: Agregar un nuevo elemento radiactivo
print("\n6. 🆕 AGREGAR NUEVO ELEMENTO (Tu ejercicio)")

# Agregar Americio-241 (usado en detectores de humo)
elementos_radiactivos["Americio-241"] = {
    "simbolo": "Am-241",
    "numero_atomico": 95,
    "masa_atomica": 241.056829,
    "vida_media": 432.2,  # años
    "tipo_decaimiento": "alfa",
    "energia_decaimiento_MeV": 5.486,
    "usos": ["detectores de humo", "densímetros"],
    "descubrimiento": 1944,
    "descubridor": "Glenn T. Seaborg"
}

print("Nuevo elemento agregado: Americio-241")

# Búsqueda de elementos por uso específico
print("\n7. 🔍 BÚSQUEDA POR APLICACIONES")
uso_buscar = "medicina nuclear"
print(f"Elementos usados en {uso_buscar}:")
for elemento, info in elementos_radiactivos.items():
    if uso_buscar in info['usos']:
        print(f"  - {elemento}")

# Análisis de masa atómica promedio
print("\n8. ⚖️ PROPIEDADES PROMEDIO")
masas = [info['masa_atomica'] for info in elementos_radiactivos.values()]
vidas_medias = [info['vida_media'] for info in elementos_radiactivos.values()]

print(f"Masa atómica promedio: {sum(masas)/len(masas):.2f} u")
print(f"Vida media promedio: {sum(vidas_medias)/len(vidas_medias):.2f} años")

# Elementos con mayor energía de decaimiento
print("\n9. 🏆 ELEMENTOS CON MAYOR ENERGÍA DE DECAIMIENTO")
elementos_ordenados = sorted(elementos_radiactivos.items(),
                            key=lambda x: x[1]['energia_decaimiento_MeV'],
                            reverse=True)

print("Top 3 elementos por energía de decaimiento:")
for i, (elemento, info) in enumerate(elementos_ordenados[:3], 1):
    print(f"{i}. {elemento}: {info['energia_decaimiento_MeV']} MeV")

print("\n¡Análisis de física nuclear completado! ⚛️")
````

````python

### Definimos los el diccionario con todos los datos a procesar.
elementos_radiactivos = {
    "Uranio-235": {
        "simbolo": "U-235",
        "numero_atomico": 92,
        "masa_atomica": 235.0439299,
        "vida_media": 703800000,  # años
        "tipo_decaimiento": "alfa",
        "energia_decaimiento_MeV": 4.679,
        "usos": ["reactores nucleares", "armas nucleares"],
        "descubrimiento": 1935,
        "descubridor": "Arthur Jeffrey Dempster"
    },
    "Plutonio-239": {
        "simbolo": "Pu-239",
        "numero_atomico": 94,
        "masa_atomica": 239.0521634,
        "vida_media": 24100,  # años
        "tipo_decaimiento": "alfa",
        "energia_decaimiento_MeV": 5.245,
        "usos": ["reactores nucleares", "armas nucleares", "generadores termoeléctricos"],
        "descubrimiento": 1941,
        "descubridor": "Glenn T. Seaborg"
    },
    "Carbono-14": {
        "simbolo": "C-14",
        "numero_atomico": 6,
        "masa_atomica": 14.003241,
        "vida_media": 5730,  # años
        "tipo_decaimiento": "beta",
        "energia_decaimiento_MeV": 0.156,
        "usos": ["datación arqueológica", "trazadores biológicos"],
        "descubrimiento": 1940,
        "descubridor": "Martin Kamen"
    },
    "Radio-226": {
        "simbolo": "Ra-226",
        "numero_atomico": 88,
        "masa_atomica": 226.0254098,
        "vida_media": 1600,  # años
        "tipo_decaimiento": "alfa",
        "energia_decaimiento_MeV": 4.870,
        "usos": ["medicina nuclear", "investigación"],
        "descubrimiento": 1898,
        "descubridor": "Marie Curie"
    },
    "Tecnecio-99m": {
        "simbolo": "Tc-99m",
        "numero_atomico": 43,
        "masa_atomica": 98.9062547,
        "vida_media": 6.0058,  # horas
        "tipo_decaimiento": "gamma",
        "energia_decaimiento_MeV": 0.141,
        "usos": ["medicina nuclear", "diagnóstico por imagen"],
        "descubrimiento": 1937,
        "descubridor": "Carlo Perrier"
    }
}

runTask = True
while runTask:
    print("==== MENU PRINCIPAL ===")
    print("1. 📊 INFORMACIÓN GENERAL DE ELEMENTOS RADIACTIVOS.")
    print("\n2. ⚡ TIPOS DE DECAIMIENTO RADIACTIVO")
    print("\n3. 🧮 CÁLCULO DE ACTIVIDAD RADIACTIVA")
    print("\n4. 📅 DESCUBRIMIENTOS POR DÉCADA")
    print("\n5. ⚡ ENERGÍA POR DECAIMIENTO")
    print("\n6. 🆕 AGREGAR NUEVO ELEMENTO (Tu ejercicio)")
    print("\n7. 🔍 BÚSQUEDA POR APLICACIONES")
    print("\n8. ⚖️ PROPIEDADES PROMEDIO")
    print("\n9. 🏆 ELEMENTOS CON MAYOR ENERGÍA DE DECAIMIENTO")
    print("\n10. SALIR")
    opcion = int(input("Ingresa el número que opción que deseas validar."))
    if opcion == 1:
        # Análisis 1: Información general de los elementos
        print("Elementos en la base de datos:")
        for elemento in elementos_radiactivos.keys():
            print(f"  - {elemento} ({elementos_radiactivos[elemento]['simbolo']})")
            
    elif opcion == 2:
        # Análisis 2: Elementos por tipo de decaimiento
        decaimientos = {}
        for elemento, info in elementos_radiactivos.items():
            tipo = info['tipo_decaimiento']
            if tipo in decaimientos:
                decaimientos[tipo].append(elemento)
            else:
                decaimientos[tipo] = [elemento]
        ### Imprime los datos de tipo de decaimento
        for tipo, elementos in decaimientos.items():
            print(f"Decaimiento {tipo}: {len(elementos)} elementos")
            for elem in elementos:
                print(f"  - {elem}")
                
    elif opcion == 3:
        # Análisis 3: Cálculo de actividad radiactiva
        print("Constante de decaimiento (λ) y actividad para 1 gramo:")
        for elemento, info in elementos_radiactivos.items():
            # λ = ln(2) / vida_media
            vida_media = info['vida_media']
            # Convertir a segundos (asumiendo que la vida media está en años)
            vida_media_segundos = vida_media * 365.25 * 24 * 3600
            constante_decaimiento = 0.693 / vida_media_segundos

            # Número de átomos en 1 gramo
            numero_atomos = 1 / info['masa_atomica'] * 6.022e23  # Número de Avogadro
            actividad = constante_decaimiento * numero_atomos  # Bq (decaimientos/segundo)
            print(f"{elemento}: λ = {constante_decaimiento:.2e} s⁻¹, Actividad = {actividad:.2e} Bq")
    
    elif opcion == 4:
        # Análisis 4: Elementos descubiertos por década
        decadas = {}
        for elemento, info in elementos_radiactivos.items():
            año = info['descubrimiento']
            decada = (año // 10) * 10
            if decada in decadas:
                decadas[decada].append(elemento)
            else:
                decadas[decada] = [elemento]
        ### Se ordena elementos por decadas (no las guarda solo las imprime.
        for decada in sorted(decadas.keys()):
            print(f"Década {decada}s: {len(decadas[decada])} elementos")
            for elem in decadas[decada]:
                print(f"  - {elem} ({elementos_radiactivos[elem]['descubridor']})")
                
    elif opcion == 5:
        # Análisis 5: Energía liberada por decaimiento
        print("Energía liberada en cada decaimiento:")
        for elemento, info in elementos_radiactivos.items():
            energia_mev = info['energia_decaimiento_MeV']
            energia_joules = energia_mev * 1.602e-13  # Conversión a joules
            print(f"{elemento}: {energia_mev} MeV = {energia_joules:.2e} J")
        
    elif opcion == 6:
        print("Agregando un elemento a la Base Datos:")
        simbolo = input("Ingresa el simbolo del nuevo elemento.")
        numero_atomico = int(input("Ingresa el número atomico del nuevo elemento: "))
        masa_atomica = float(input("Ingresa la masa atomica del nuevo elemento: "))
        vida_media = float(input("Ingresa la visa media del nuevo elemento: "))
        tipo_decaimiento = input("Ingresa el tipo de decaimiento del nuevo elemento: ")
        energia_decaimiento_MeV = input("Ingresa la energia decaimiento del nuevo elemento: ")
        descubrimiento = int(input("Ingresa el año de descubrimiento del nuevo elemento: "))
        usos = []
        while uso != "-1":
            uso = input("Ingresa un uso o -1 para terminar: ")
            usos.append(uso)
        descubridor = input("Ingresa el nombre del descubridor del nuevo elemento: ")
        nuevo_elemento = {
            "simbolo": simbolo,
            "numero_atomico": numero_atomico,
            "masa_atomica": masa_atomica,
            "vida_media": vida_media,  # años
            "tipo_decaimiento": tipo_decaimiento,
            "energia_decaimiento_MeV": energia_decaimiento_MeV,
            "usos": usos,
            "descubrimiento": descubrimiento,
            "descubridor":descubridor
        }
        elementos_radiactivos.append(nuevo_elemento)
        print("Nuevo elemento agregado")

    elif opcion == 8:
        # Búsqueda de elementos por uso específico
        uso_buscar = "medicina nuclear"
        print(f"Elementos usados en {uso_buscar}:")
        for elemento, info in elementos_radiactivos.items():
            if uso_buscar in info['usos']:
                print(f"  - {elemento}")
    
    elif opcion == 9:
        # Análisis de masa atómica promedio
        masas = [info['masa_atomica'] for info in elementos_radiactivos.values()]
        vidas_medias = [info['vida_media'] for info in elementos_radiactivos.values()]

        print(f"Masa atómica promedio: {sum(masas)/len(masas):.2f} u")
        print(f"Vida media promedio: {sum(vidas_medias)/len(vidas_medias):.2f} años")
        # Elementos con mayor energía de decaimiento

        elementos_ordenados = sorted(elementos_radiactivos.items(),
                                    key=lambda x: x[1]['energia_decaimiento_MeV'],
                                    reverse=True)

        print("Top 3 elementos por energía de decaimiento:")
        for i, (elemento, info) in enumerate(elementos_ordenados[:3], 1):
            print(f"{i}. {elemento}: {info['energia_decaimiento_MeV']} MeV")

        print("\n¡Análisis de física nuclear completado! ⚛️")

    elif opcion == 10:
        print("SALIENDO DEL PROGRAMA")
        runTask = False
    else:
        print ("opcion invalida")
````

### Ejercicio 1: Sistema de Registro de Pacientes
Crea un programa que permita gestionar el registro de pacientes de una clínica usando diccionarios.

**Requisitos:**
1. Crea un diccionario vacío llamado `registro_pacientes`
2. Permite al usuario agregar 3 pacientes con la siguiente información:
   - Nombre (clave)
   - Edad (valor: número)
   - Enfermedad (valor: string)
   - Fecha de ingreso (valor: string en formato "dd/mm/aaaa")
3. Muestra toda la información de los pacientes registrados
4. Permite buscar un paciente por nombre y mostrar sus datos
5. Calcula y muestra el promedio de edad de los pacientes

````python
### Definimos los el diccionario con todos los datos a procesar.
lista_registro_pacientes = []

runTask = True
while runTask:
    print("==== MENU PRINCIPAL ===")
    print("1. AGREGAR NUEVO ELEMENTO")
    print("\n2. MOSTRAR TODOS LOS PACIENTES")
    print("\n3. BÚSQUEDA POR NOMBRE")
    print("\n4. PROMEDIO DE PACIENTES")
    print("\n5. SALIR")
    opcion = int(input("Ingresa el número que opción que deseas validar."))

    if opcion == 1:
        ### 1. Agregando nueva información del paciente.
        print("Agregando un elemento a la Base Datos:")
        nombre = input("Ingresa el nombre del pacientea : ")
        edad = int(input("Ingresa la edad del paciente: "))
        enfermedad = input("Ingresa la enfermedad del paciente: ")
        fecha_ingreso = input("Ingresa la fecha de ingreso en dd/mm/aaaa: ")
        registro_pacientes = {
            "nombre": nombre,
            "edad": edad,
            "enfermedad": enfermedad,
            "fecha_ingreso": fecha_ingreso
        }
        lista_registro_pacientes.append(registro_pacientes)
        print("Nuevo elemento agregado")
    
    elif opcion == 2:
        # 2. Información general de los elementos
        print("Elementos en la base de datos")
        for i, paciente in enumerate(lista_registro_pacientes):
            print(f" [{i}] : ({paciente})")
            
    elif opcion == 3:
        # 3. Información general a buscar
        print("Elementos en la base de datos")
        nombre = input("Ingresa el nombre del pacientea : ")
        for i, paciente in enumerate(lista_registro_pacientes):
            if paciente["nombre"] == nombre:
                print(f" [{i}] : ({paciente})")
                break
                
    elif opcion == 4:
        # 4. Promedio
        print("Elementos en la base de datos")
        promedio = 0
        for paciente in lista_registro_pacientes:
            promedio = paciente["edad"]/len(lista_registro_pacientes)
        print("PROMEDIO: ", promedio)
    elif opcion == 5:
        print("SALIENDO DEL PROGRAMA")
        runTask = False
    else:
        print ("opcion invalida")
````

### Ejercicio 2: 🎬 Base de Datos de Películas

Crea un sistema para gestionar una colección de películas usando diccionarios anidados.

**Requisitos:**
1. Crea un diccionario llamado `coleccion_peliculas` con al menos 4 películas
2. Cada película debe tener:
   - Título (clave)
   - Año (valor: número)
   - Género (valor: string)
   - Duración en minutos (valor: número)
   - Calificación (valor: número del 1 al 5)
3. Implementa las siguientes funciones:
   - Mostrar todas las películas
   - Buscar películas por género
   - Encontrar la película más larga
   - Calcular el promedio de calificaciones

````python
### Definimos los el diccionario con todos los datos a procesar.
coleccion_peliculas = {
    "Shrek": {
        "año": 2002,
        "genero": "Comedia",
        "duracion": 65.04,
        "calificacion": 5
    },
    "Indiana Jones": {
        "año": 1982,
        "genero": "Aventura",
        "duracion": 71.43,
        "calificacion": 4
    },
    "Iron Man": {
        "año": 2008,
        "genero": "Sci-Fi",
        "duracion": 85.13,
        "calificacion": 5
    },
    "La Era del Hielo": {
        "año": 2001,
        "genero": "Comedia",
        "duracion": 91.61,
        "calificacion": 4
    },
    "El Conjuro": {
        "año": 2015,
        "genero": "Terror",
        "duracion": 129.74,
        "calificacion": 3
    },
}

runTask = True
while runTask:
    print("==== MENU PRINCIPAL ===")
    print("1. Mostrar todas las películas")
    print("\n2. Buscar películas por género ")
    print("\n3. Encontrar la película más larga")
    print("\n4. Calcular el promedio de calificaciones")
    print("\n5. Salir")
    opcion = int(input("Ingresa el número que opción que deseas validar."))

    
    if opcion == 1:
        ### 1. Información general.
        print("Elementos en la base de datos")
        for i, pelicula in enumerate(coleccion_peliculas.keys()):
            print(f"{i}.[{pelicula}] : ({coleccion_peliculas[pelicula]})")
            
    elif opcion == 2:
        ### 2. Buscar por categoría.
        cat = input("Ingresa el la categoría que deseas buscar: ")
        for i, pelicula in enumerate(coleccion_peliculas.keys()):
            if coleccion_peliculas[pelicula]["genero"] == cat:
                print(f"{i}.[{pelicula}] : ({coleccion_peliculas[pelicula]})")
                             
    elif opcion == 3:
        ### 3. Saca la pelicula con más duración.
        max_duracion = 0
        for i, pelicula in enumerate(coleccion_peliculas.keys()):
            if coleccion_peliculas[pelicula]["duracion"] >= max_duracion:
                max_duracion = coleccion_peliculas[pelicula]["duracion"]
                key = pelicula
        
        print(f"Pelicula con mayor duracion: [{key}] : ({coleccion_peliculas[key]})")
    elif opcion == 4:
        ### 4. Promedio de duraciones
        promedio = 0
        for pelicula in coleccion_peliculas.keys():
            promedio += coleccion_peliculas[pelicula]["duracion"]/len(coleccion_peliculas.keys())
        print("PROMEDIO: ", promedio)
    elif opcion == 5:
        print("SALIENDO DEL PROGRAMA")
        runTask = False
    else:
        print ("opcion invalida")
````
