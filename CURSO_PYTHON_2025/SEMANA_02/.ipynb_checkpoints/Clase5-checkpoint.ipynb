## Introducción a los bucles

Los bucles son estructuras de control que nos permiten repetir un bloque de código múltiples veces. En Python, el bucle `for` es la principal herramienta para iterar sobre elementos de una secuencia.

**¿Por qué usar bucles?**
- Automatizar tareas repetitivas
- Procesar colecciones de datos
- Evitar código repetitivo
- Hacer nuestro código más eficiente y legible

```python
# Ejemplo básico de bucle for
print("=== BUCLE FOR BÁSICO ===")

# Iterar sobre una lista de frutas
frutas = ["manzana", "banana", "cereza", "naranja"]

i=0
print("Lista de frutas con while:")
while i<4:
  print(f" - {frutas[i]}")
  i+=1

print("\nLista de frutas con for:")
for fruta in frutas:
    print(f" - {fruta}")

```

```python
# Iterar sobre una cadena de texto
print("\nLetras en 'Python':")
python_arreglo = ["P","y","t","h","o","n"]
for letra in "Python":
    print(f" - {letra}")

# Contar elementos (forma tradicional)
print("\nContando frutas:")
contador = 1
for fruta in frutas:
    print(f"{contador}. {fruta}")
    contador += 1
```

### La función range()

La función `range()` es fundamental para crear secuencias numéricas que podemos usar en bucles for. Tiene tres formas principales:

- `range(stop)`: crea una secuencia de 0 a stop-1
- `range(start, stop)`: crea una secuencia de start a stop-1  
- `range(start, stop, step)`: crea una secuencia con paso específico

```python
# Uso de la función range()
print("=== FUNCIÓN RANGE() ===")

# range(stop)
print("range(5):")
for i in range(5):
    print(f" - {i}")

# range(start, stop)
print("\nrange(2, 7):")
for i in range(2, 7):
    print(f" - {i}")

# range(start, stop, step)
print("\nrange(0, 10, 2):")
for i in range(0, 10, 2):
    print(f" - {i}")

# Ejemplo práctico: tabla de multiplicar
print("\nTabla del 5:")
for i in range(1, 11):
    print(f"5 × {i} = {5 * i}")
```

### Iterando sobre diferentes estructuras

Python nos permite iterar sobre diferentes tipos de estructuras de datos:

```python
# Iterando sobre diferentes estructuras de datos
print("=== ITERANDO SOBRE DIFERENTES ESTRUCTURAS ===")

# Listas
print("Lista:")
numeros = [10, 20, 30, 40]
for numero in numeros:
    print(f" - {numero}")

# Tuplas
print("\nTupla:")
coordenadas = (3, 5, 7)
for coord in coordenadas:
    print(f" - {coord}")

# Conjuntos
print("\nConjunto:")
colores = {"rojo", "verde", "azul"}
for color in colores:
    print(f" - {color}")
```

### La función enumerate()

La función `enumerate()` es muy útil cuando necesitamos tanto el elemento como su índice durante la iteración. Devuelve pares (índice, elemento).

```python
# Uso de enumerate()
print("=== FUNCIÓN ENUMERATE() ===")

frutas = ["manzana", "banana", "cereza", "naranja"]

print("Enumerando frutas:")
for indice, fruta in enumerate(frutas):
    print(f"{indice}: {fruta}")

# Podemos especificar el índice inicial
print("\nEnumerando con inicio personalizado:")
for indice, fruta in enumerate(frutas, start=1):
    print(f"{indice}. {fruta}")

# Ejemplo práctico: buscar un elemento
print("\nBuscando 'cereza':")
buscar = "banna"
for indice, fruta in enumerate(frutas,start=1):
    if fruta == buscar:
        print(f"¡Encontrado! '{buscar}' está en la posición {indice}")
        break
else:
    print(f"'{buscar}' no encontrado")
```

### La función zip()

La función `zip()` nos permite iterar sobre múltiples secuencias simultáneamente, combinando elementos en tuplas.

```python
# Uso de zip()
print("=== FUNCIÓN ZIP() ===")

nombres = ["Ana", "Luis", "Carlos"]
edades = [25, 30, 22]
ciudades = ["Madrid", "Barcelona", "Valencia"]

print("Combinando listas con zip():")
for nombre, edad, ciudad in zip(nombres, edades, ciudades):
    print(f"{nombre} tiene {edad} años y vive en {ciudad}")

# Zip con diferente longitud
print("\nZip con diferentes longitudes:")
colores = ["rojo", "verde", "azul", "amarillo"]  # 4 elementos
codigos = [100, 200, 300]                        # 3 elementos

for color, codigo in zip(colores, codigos):
    print(f"Código {codigo}: {color}")

# Convertir a lista para ver el resultado
print(f"\nResultado de zip(): {list(zip(colores, codigos))}")
```

### Sentencias break, continue y else

Python ofrece varias sentencias para controlar el flujo de los bucles:

- `break`: Termina el bucle completamente
- `continue`: Salta a la siguiente iteración
- `else`: Se ejecuta si el bucle termina normalmente (sin break)

```python
# Break, continue y else
print("=== BREAK, CONTINUE Y ELSE ===")

# Break: terminar el bucle anticipadamente
print("Uso de break (buscar el número 5):")
for numero in range(10):
    if numero == 5:
        print("¡Número 5 encontrado! Terminando bucle...")
        break
    print(f"Número actual: {numero}")

# Continue: saltar a la siguiente iteración
print("\nUso de continue (saltar números pares):")
for numero in range(10):
    if numero % 2 == 0:  # Si es par
        continue         # Saltar a la siguiente iteración
    print(f"Número impar: {numero}")

# Else: ejecutar si no hubo break
print("\nUso de else (buscar número que no existe):")
buscar = 4
for numero in range(10):
    if numero == buscar:
        print(f"¡{buscar} encontrado!")
        break
else:
 print(f"{buscar} no encontrado en el rango 0-9")
```

### Comprensión de listas

La comprensión de listas es una forma concisa de crear listas aplicando operaciones a cada elemento de una secuencia. Es como un bucle for compacto.

```python
# Comprensión de listas
print("=== COMPRENSIÓN DE LISTAS ===")

# Forma tradicional
print("Cuadrados (forma tradicional):")
cuadrados = []
for x in range(10):
    cuadrados.append(x**2)
print(cuadrados)

# Forma con comprensión de listas
print("\nCuadrados (comprensión de listas):")
cuadrados = [x**2 for x in range(10)]
print(cuadrados)

# Comprensión con condición
print("\nNúmeros pares al cuadrado:")
pares_cuadrados = [x**2 for x in range(10) if x % 2 == 0]
print(pares_cuadrados)

# Comprensión con if-else
print("\nPar o impar:")
paridad = ["par" if x % 2 == 0 else "impar" for x in range(10)]
print(paridad)

# Comprensión con múltiples bucles
print("\nCombinaciones:")
combinaciones = [(x, y) for x in range(3) for y in range(3)]
print(combinaciones)
```

### Bucles anidados

Podemos colocar bucles dentro de otros bucles para trabajar con estructuras multidimensionales.

```python
# Bucles anidados
print("=== BUCLES ANIDADOS ===")

# Matriz 2D
matriz = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]

print("Recorriendo matriz:")
for i, fila in enumerate(matriz):
    for j, valor in enumerate(fila):
        print(f"matriz[{i}][{j}] = {valor}")

# Tabla de multiplicar
print("\nTablas de multiplicar:")
for i in range(1, 11):  # Tablas del 1 al 3
    print(f"\nTabla del {i}:")
    for j in range(1, 11):  # Multiplicar del 1 al 5
        print(f"{i} × {j} = {i * j}")
```

## Ejercicios de ejemplo

```python
# Ejercicios prácticos
print("=== EJERCICIOS PRÁCTICOS ===")

# Ejercicio 1: Contar vocales en una frase
print("Ejercicio 1: Contar vocales")
frase = "Python es un lenguaje de programación"
vocales = "aeiouAEIOU"
contador = 0

for letra in frase:
    if letra in vocales:
        contador += 1

print(f"Frase: '{frase}'")
print(f"Número de vocales: {contador}")

# Ejercicio 2: Encontrar el número más grande
print("\nEjercicio 2: Encontrar el máximo")
numeros = [34, 12, 89, 5, 67, 23]
maximo = numeros[0]

for numero in numeros:
    if numero > maximo:
        maximo = numero

print(f"Lista: {numeros}")
print(f"Valor máximo: {maximo}")

# Ejercicio 3: Filtrar palabras por longitud
print("\nEjercicio 3: Filtrar palabras largas")
palabras = ["casa", "programación", "sol", "computadora", "kiwi"]
longitud_minima = 5
palabras_largas = []

for palabra in palabras:
    if len(palabra) >= longitud_minima:
        palabras_largas.append(palabra)

print(f"Palabras: {palabras}")
print(f"Palabras con {longitud_minima}+ letras: {palabras_largas}")

# Ejercicio 4: Crear un diccionario de frecuencias
print("\nEjercicio 4: Diccionario de frecuencias")
texto = "hola mundo hola python mundo hola"
palabras = texto.split()
frecuencias = {}

for palabra in palabras:
    if palabra in frecuencias:
        frecuencias[palabra] += 1
    else:
        frecuencias[palabra] = 1

print(f"Texto: '{texto}'")
print(f"Frecuencias: {frecuencias}")
```

##  Ejercicios de práctica

A continuación, te presento 3 ejercicios para practicar lo aprendido sobre bucles for. Intenta resolverlos por tu cuenta antes de ver las soluciones.

```python
# Ejercicio 1: Suma de números pares
print("=== EJERCICIO 1: SUMA DE NÚMEROS PARES ===")
print("Escribe un programa que sume todos los números pares del 1 al 100.")

### Generamos nuestra lista
lista = list(range(1,101,1))
### Sumanos todos los valores
par = 0
i=0
while i < 100:
    if (lista[i]%2) == 0:
        par+=lista[i]
    i+=1
print(f"La suma de todos los Pares del 1 al 100 es: {par}")
```

```python
# Ejercicio 2: Contador de vocales
print("\n=== EJERCICIO 2: CONTADOR DE VOCALES ===")
print("Escribe un programa que cuente cuántas vocales tiene una palabra ingresada por el usuario.")

### Variables
a = 0
e = 0
i = 0
o = 0
u = 0
### Pedimos la palabra
palabra = list(input("Ingrese la palabra que desea procesar: "))
for letra in palabra:
    if letra == 'a' or letra == 'A':
        a+=1
    elif letra == 'e' or letra == 'E':
        e+=1
    elif letra == 'i' or letra == 'I':
        i+=1
    elif letra == 'o' or letra == 'O':
        o+=1
    elif letra == 'u' or letra == 'U':
        u+=1
print(f"A: {a}, E: {e}, I: {i}, O: {o}, U: {u}")
```
